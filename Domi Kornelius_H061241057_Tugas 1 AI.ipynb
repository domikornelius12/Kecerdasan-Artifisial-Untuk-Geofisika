**# TUGAS INDIVIDU AI GEOFISIKA**
**Topik:** Perancangan Prompt Spesifikasi Kerja (*Prompt Work Specification*) untuk Studi Kasus Geofisika  
**Kerangka Acuan:** Struktur Prompt *AI Cookbook untuk Earth Sciences* (Gambar 2.1: R - T - C - I - K - O - Q - V)

---

**## Modul Unsur-Unsur Prompt**
1. **R (Role):** Peran spesifik AI (misal: Ahli Geofisika, Konsultan Kebencanaan, Data Analyst).
2. **T (Task):** Tugas utama yang harus diselesaikan.
3. **C (Context):** Latar belakang geologi, lokasi, atau kondisi pengukuran.
4. **I (Input):** Data mentah, tabel, parameter, atau asumsi awal yang disediakan.
5. **K (Constraints):** Batasan teknis, teori geofisika yang tidak boleh dilanggar, atau aturan penulisan.
6. **O (Output):** Format hasil yang diinginkan (laporan singkat, narasi deskriptif, tabel, atau skema visual).
7. **Q (Quality Criteria):** Standar kelayakan scientific/geofisika agar output dapat diterima.
8. **V (Optional Verification):** Langkah verifikasi silang atau logika pembanding.

**# 1. STUDI KASUS GEOTHERMAL: DELINEASI REBOISASI DAN RESERVOIR DENGAN MAGNETOTELLURIK (MT)**

* **Role**
  `Anda berperan sebagai seorang Geothermal Geophysicist Senior yang memiliki keahlian mendalam dalam pemodelan data Magnetotellurik (MT) 2D/3D, analisis fasa dan impedansi sistem panas bumi, serta integrasi hidrotermal pada sistem vulkanik aktif.`

* **Task**
  `Lakukan analisis dan interpretasi komprehensif terhadap penampang resistivitas 2D hasil inversi Magnetotellurik (MT) untuk mendelineasi batas-batas sistem panas bumi. Analisis harus mencakup identifikasi ketebalan dan sebaran lapisan penudung (clay cap), batas atas dan samping dari zona reservoir, perkiraan kedalaman sumber panas (heat source), serta penentuan area permeabel yang berpotensi memiliki konektivitas fluida tinggi.`

* **Context**
  `Evaluasi ini dilakukan pada area lapangan panas bumi prospektif di lingkungan gunung api kuarter yang sedang direncanakan untuk tahap pemboran eksplorasi. Pemerintah dan pihak pengembang membutuhkan kepastian mengenai struktur konseptual bawah permukaan guna meminimalkan risiko kegagalan pemboran (dry hole) serta menentukan kedalaman target sumur yang paling optimal.`

* **Input**
  `Data penampang 2D resistivitas hasil inversi MT (frekuensi 0.001 Hz - 1000 Hz), rentang nilai resistivitas konduktif (< 10 ohm.m), resistivitas sedang (30 - 100 ohm.m), dan resistivitas tinggi (> 200 ohm.m), peta geologi permukaan lokal yang menunjukkan zona ubahan hidrotermal (argilik dan propilitik), data manifes permukaan (fumarol dan mata air panas asam/netral), serta data penginderaan jauh terkait struktur kelurusan (lineament).`

* **Constraints**
  `Dalam melakukan interpretasi, jangan secara langsung menyimpulkan bahwa zona dengan resistivitas paling konduktif (< 10 ohm.m) merupakan zona reservoir panas bumi, melainkan harus diinterpretasikan sebagai lapisan penudung (clay cap) berkarakteristik keterubahan mineral smektit. Interpretasi zona reservoir harus dikaitkan dengan zona resistivitas sedang di bawah clay cap yang berasosiasi dengan mineral ubahan temperatur lebih tinggi (illit/klorit). Hindari memberikan estimasi temperatur absolut jika tidak didukung oleh data geotermometer fluida.`

* **Output**
  `Rencana output berupa narasi laporan evaluasi geofisika sistem panas bumi yang disusun secara sistematis, meliputi:`
    `1. Analisis karakteristik tahanan jenis dan delineasi ketebalan serta variasi kedalaman zona clay cap.`
    `2. Pemodelan konseptual geometri reservoir mencakup estimasi kedalaman top of reservoir dan batas lateralnya.`
    `3. Diskusi mengenai hubungan antara kontrol struktur geologi (patahan/rekahan) dengan zona permeabilitas tempat fluida naik.`
    `4. Evaluasi faktor ketidakpastian inversi data MT pada kedalaman dangkal dan sangat dalam.`
    `5. Rekomendasi lokasi dan target kedalaman pemboran sumur landaian suhu atau sumur eksplorasi.`

**# 2. STUDI KASUS PERTAMBANGAN: EKSPLORASI BIJIH BESI DENGAN METODE GEOMAGNETIK**

* **Role**
  `Anda berperan sebagai seorang Mining Geophysicist ahli yang berpengalaman dalam akuisisi, pengolahan data medan magnetik bumi, aplikasi pemrosesan sinyal (filter penataan/transformasi medan), serta pemodelan geometri tubuh bijih kaya mineral magnetit.`

* **Task**
  `Lakukan evaluasi dan interpretasi menyeluruh terhadap data anomali magnetik total beserta hasil filter pengolahan citra guna memetakan sebaran lateral, kedalaman puncak, dan estimasi kemiringan (dip) dari tubuh bijih besi (magnetit). Anda diminta untuk memisahkan respon anomali magnetik yang bersumber dari mineralisasi utama dengan respon anomali latar belakang akibat batuan samping.`

* **Context**
  `Survei magnetik darat dilakukan pada suatu wilayah prospek pertambangan bijih besi yang sebagian besar permukaannya tertutup oleh endapan aluvial dan tanah pelapukan (overburden) yang tebal. Informasi mengenai batas bawah permukaan dari zona mineralisasi sangat dibutuhkan oleh tim geologi tambang untuk merencanakan titik-titik pemboran inti (core drilling) secara efisien.`

* **Input**
  `Peta Anomali Magnetik Total (Total Magnetic Anomaly - TMA), peta hasil transformasi Reduce to Pole (RTP), peta Analytic Signal, peta First Vertical Derivative (1VD), data pengukuran suseptibilitas magnetik dari sampel batuan inti/singkapan lokal, serta data ketebalan tanah penutup berdasarkan beberapa data parit uji (trenching).`

* **Constraints**
  `Analisis harus memperhitungkan efek dari kemagnetan sisa (remanent magnetization) yang berpotensi menyebabkan pergeseran posisi anomali puncak pada peta RTP. Jangan mengasumsikan seluruh nilai anomali magnetik tinggi sebagai tubuh bijih murni tanpa mempertimbangkan variasi suseptibilitas batuan samping (seperti batuan basal atau andesit). Wajib menjelaskan ambiguitas pemodelan magnetik yang bersifat non-unique.`

* **Output**
  `Rencana output berupa narasi ilmiah-teknis yang terstruktur, memuat:`
  `1. Penjelasan prinsip fisik respon magnetik terhadap kandungan mineral magnetit.`
  `2. Analisis perbandingan peta TMA, RTP, dan Analytic Signal dalam menentukan pusat kemagnetan tubuh bijih.`
  `3. Interpretasi struktur geologi pengontrol mineralisasi (seperti zona sesar atau kontak batuan) berdasarkan peta derivatif (1VD).`
  `4. Perkiraan dimensi geometri, kedalaman puncak (top depth), dan arah kemiringan tubuh mineralisasi.`
  `5. Penilaian tingkat ketidakpastian data dan rekomendasi koordinat titik bor verifikasi.`

**# 3. STUDI KASUS MITIGASI BENCANA: PEMETAAN ZONA RENTAN LIKUIFAKSI MENGGUNAKAN METODE SEISMIK REFRAKSI**

* **Role**
  `Anda berperan sebagai seorang Geophysicist Bencana dan Geoteknik yang memiliki spesialisasi dalam bidang analisis dinamika tanah, pemrosesan sinyal seismik refraksi, tomografi gelombang kompresi (Vp), serta penilaian risiko kebencanaan akibat guncangan gempa bumi.`

* **Task**
  `Menganalisis profil kecepatan gelombang kompresi 2D (Vp) hasil tomografi refraksi seismik untuk memetakan variasi ketebalan sedimen lunak, kedalaman lapisan batuan dasar (bedrock), serta mengidentifikasi zona perlapisan pasir lepas jenuh air yang memiliki kerentanan tinggi terhadap fenomena likuifaksi.`

* **Context**
  `Investigasi geofisika dangkal ini dilakukan di kawasan pemukiman padat penduduk yang berada dekat dengan zona sesar aktif. Wilayah ini berdiri di atas endapan aluvial dan pernah mengalami guncangan gempa. Pemerintah daerah memerlukan peta zonasi kerentanan tanah sebagai dasar penyusunan tata ruang dan standar keteknikan fondasi bangunan baru.`

* **Input**
  `Data waktu tempuh gelombang pertama (first-break travel time), penampang profil kecepatan Vp 2D hasil inversi tomografi, data variasi kedalaman muka air tanah dangkal hasil pengukuran sumur warga (1 - 4 meter), deskripsi sampel pemboran geoteknik dangkal, serta data percepatan batuan dasar rencana.`

* **Constraints**
  `Dalam menentukan potensi likuifaksi, jangan hanya mengandalkan satu nilai Vp semata, tetapi wajib mengintegrasikan analisis antara nilai Vp rendah hingga sedang (300 m/s - 1200 m/s), tingkat kejenuhan air (di mana Vp pada sedimen jenuh air umumnya melonjak mendekati 1500 m/s akibat sifat fluida), dan keberadaan muka air tanah dangkal. Jelaskan secara tegas keterbatasan seismik refraksi dalam membedakan batas antar-lapisan jika terjadi kondisi Vp yang mengalami velocity reversal (lapisan lambat di bawah lapisan cepat).`

* **Output**
  `Rencana output berupa narasi laporan komprehensif mitigasi bencana, meliputi:`
  `1. Interpretasi pola distribusi kecepatan gelombang Vp terhadap kedalaman dan jarak lintasan.`
  `2. Pemetaan ketebalan lapisan sedimen permukaan dan delineasi profil kedalaman bedrock.`
  `3. Analisis tumpang-tindih (overlay) antara zona Vp jenuh air dengan kriteria material pasir lepas rawan likuifaksi.`
  `4. Diskusi mengenai faktor ketidakpastian tomografi refraksi dan batasan resolusi lateral.`
  `5. Kesimpulan kondisi tapak (site condition) serta rekomendasi pengujian geoteknik lanjutan (seperti CPT atau SPT).`

**# 4. STUDI KASUS KETEKNIKAN: DETEKSI RONGGA BAWAH PERMUKAAN (KARST) DENGAN GROUND PENETRATING RADAR (GPR)**

* **Role**
  `Anda berperan sebagai seorang Near-Surface Geophysicist yang menguasai teori propagasi gelombang elektromagnetik (EM), pemrosesan sinyal radar, interpretasi radargram, dan investigasi geofisika untuk kebutuhan infrastruktur transportasi.`

* **Task**
  `Analisis dan interpretasikan data rekaman profil radargram GPR 2D untuk mendeteksi lokasi, kedalaman, serta perkiraan dimensi rongga atau gua karst (void/sinkhole) di bawah permukaan yang berpotensi memicu amblesan atau runtuhan tanah.`

* **Context**
  `Survei GPR ini dilaksanakan di sepanjang koridor perencanaan jalur rel kereta api berkecepatan tinggi yang melintasi kawasan bentang alam karst (batu gamping). Deteksi keberadaan rongga bawah permukaan sangat krusial dilakukan sebelum konstruksi dimulai untuk mencegah risiko kegagalan struktur fondasi rel atau bahaya kecelakaan di masa depan.`

* **Input**
  `Profil radargram GPR 2D (menggunakan frekuensi antena pusat 250 MHz), estimasi kecepatan propagasi gelombang EM pada batu gamping padat (0.1 m/ns), data pola refraksi/refleksi hiperbola, data filter pengolahan (seperti bandpass filter, background removal, dan gain correction), serta peta topografi dan elevasi permukaan.`

* **Constraints**
  `Dilarang menyimpulkan setiap anomali hiperbola pada radargram sebagai rongga karst tanpa memperhatikan fase refleksi (kontras permitivitas relatif). Pertimbangkan efek atenuasi sinyal EM yang tinggi apabila rongga terisi oleh material lempung jenuh atau air asin, yang dapat menyebabkan hilangnya penetrasi sinyal radar (radar signal loss). Bedakan dengan jelas antara respon rongga alami dengan utilitas buatan manusia (seperti pipa atau gorong-gorong).`

* **Output**
  `Rencana output berupa narasi teknis evaluasi geofisika keteknikan yang berisi:`
  `1. Penjelasan mekanisme respon gelombang EM terhadap kontras dielektrik antara batuan gamping dan rongga (berisi udara/air/sebagian lempung).`
  `2. Pemetaan posisi horizontal (jarak lintasan) dan kalkulasi kedalaman anomali hiperbola.`
  `3. Estimasi dimensi rongga berdasarkan geometri kurva hiperbola dan pola refleksi batas bawah rongga.`
  `4. Analisis kedalaman penetrasi efektif sinyal GPR serta faktor pembatas kualitas data.`
  `5. Rekomendasi teknis mitigasi geoteknik (seperti metode injeksi semen/grouting) dan verifikasi drilling atau borehole camera.`

**# 5. STUDI KASUS HIDRO-GEOLOGI LEPAS PANTAI: PEMETAAN AIR TANAH LEPAS PANTAI (OFFSHORE AQUIFER) DENGAN METODE ELECTROMAGNETIC (CSEM)**

* **Role**
  `Anda berperan sebagai seorang Marine Geophysicist Senior yang ahli dalam metode Controlled-Source Electromagnetic (CSEM) bawah laut, pemodelan inversi konduktivitas/resistivitas media fluida laut, serta hidrogeologi kawasan pesisir dan lepas pantai.`

* **Task**
  `Lakukan analisis dan evaluasi terhadap data resistivitas 2D/3D hasil pemodelan CSEM dasar laut untuk mengidentifikasi keberadaan, geometri, dan perkiraan batas sebaran lateral serta vertikal dari akuifer air tawar bawah laut (offshore freshened groundwater).`

* **Context**
  `Penelitian dilakukan di kawasan pesisir pulau kecil yang mengalami krisis air bersih dan penurunan muka air tanah yang parah akibat eksploitasi berlebih. Pengembang dan pemerintah daerah berupaya mencari sumber daya air tawar non-konvensional yang tersimpan di dalam lapisan sedimen di bawah dasar laut (lepas pantai) sebagai cadangan strategis masa depan.`

* **Input**
  `Data komponen medan listrik dan magnetik hasil perekaman transmitter-receiver CSEM di dasar laut, profil penampang inversi resistivitas bawah dasar laut (kedalaman 0 - 500 meter di bawah sediment-water interface), data kedalaman air laut (batimetri), data nilai resistivitas air laut lokal (0.3 ohm.m), serta data seismik dangkal pencitraan struktur sedimen jika tersedia.`

* **Constraints**
  `Dalam analisis, perhatikan dengan teliti bahwa air laut dan sedimen jenuh air laut memiliki nilai resistivitas yang sangat rendah (< 1 ohm.m). Oleh karena itu, keberadaan akuifer air tawar akan ditandai oleh kontras resistivitas yang lebih tinggi (biasanya > 5 - 10 ohm.m). Namun, jangan mengabaikan kemungkinan bahwa respon resistivitas tinggi tersebut berasal dari batuan karbonat, evaporit, atau lapisan gas dangkal (shallow gas). Analisis wajib memisahkan respon akibat jenis fluida dengan jenis batuan matrik.`

* **Output**
  `Rencana output berupa narasi analisis geofisika kelautan yang detail dan ilmiah, memuat:`
  `1. Prinsip kerja difusi gelombang EM di media air laut dan penetrasinya ke dalam sedimen dasar laut.`
  `2. Identifikasi dan pemetaan batas zona resistivitas tinggi yang diindikasikan sebagai akuifer air tawar lepas pantai.`
  `3. Estimasi kedalaman perlapisan akuifer di bawah dasar laut beserta perkiraan ketebalannya.`
  `4. Diskusi mendalam mengenai ambiguitas inversi CSEM, efek air-wave, dan perbedaan respon fluida vs litologi.`
  `5. Kesimpulan mengenai potensi cadangan air serta saran investigasi lanjutan melalui pemboran geotechnical offshore dan pengujian sampel geokimia fluida.`
