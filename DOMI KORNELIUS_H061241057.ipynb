```python
# ============================================================
# TUGAS KECERDASAN ARTIFISIAL UNTUK GEOFISIKA
# RECIPE R-VIS-001
# MEMBUAT TIME-SERIES FIGURE YANG INFORMATIF
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# ============================================================
# 1. PENGATURAN PARAMETER
# ============================================================

OUTPUT_DIR = Path("hasil_R_VIS_001")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

dt = 0.002
durasi = 10.0
waktu = np.arange(0, durasi, dt)

t_kedatangan = 3.0
frekuensi = 5.0

# Seed agar hasil dapat direproduksi
rng = np.random.default_rng(42)

# ============================================================
# 2. MEMBANGUN DATA SINTETIS
# ============================================================

# Membuat noise acak
noise = 0.08 * rng.normal(size=len(waktu))

# Membuat gelombang seismik sintetis
waktu_rel = waktu - t_kedatangan
gelombang = np.zeros_like(waktu)

mask = waktu >= t_kedatangan

gelombang[mask] = (
    1.2
    * np.sin(2 * np.pi * frekuensi * waktu_rel[mask])
    * np.exp(-waktu_rel[mask] / 2.0)
)

# Menggabungkan noise dan gelombang
amplitudo = noise + gelombang

# ============================================================
# 3. MENAMPILKAN INFORMASI DATA
# ============================================================

print("=" * 55)
print("PRAKTIK RECIPE R-VIS-001")
print("TIME-SERIES FIGURE UNTUK DATA GEOFISIKA")
print("=" * 55)

print(f"Jumlah sampel       : {len(waktu)}")
print(f"Interval sampling   : {dt:.3f} sekon")
print(f"Durasi pengamatan   : {durasi:.1f} sekon")
print(f"Waktu impuls        : {t_kedatangan:.1f} sekon")
print(f"Frekuensi gelombang : {frekuensi:.1f} Hz")
print(f"Amplitudo maksimum  : {np.max(amplitudo):.3f}")
print(f"Amplitudo minimum   : {np.min(amplitudo):.3f}")

# ============================================================
# 4. MEMBUAT GRAFIK TIME SERIES
# ============================================================

fig, ax = plt.subplots(figsize=(12, 5.5))

ax.plot(
    waktu,
    amplitudo,
    color="black",
    linewidth=0.8,
    label="Sinyal sintetis"
)

# Menandai waktu kedatangan impuls
ax.axvline(
    x=t_kedatangan,
    color="red",
    linestyle="--",
    linewidth=1.4,
    label=f"Waktu impuls ({t_kedatangan:.1f} s)"
)

ax.set_title(
    "Deret Waktu Sinyal Seismik Sintetis",
    fontsize=14,
    pad=12
)

ax.set_xlabel("Waktu (s)", fontsize=11)
ax.set_ylabel("Amplitudo relatif", fontsize=11)

ax.set_xlim(0, durasi)
ax.grid(True, linestyle=":", linewidth=0.7, alpha=0.8)
ax.legend(loc="upper right")

fig.tight_layout()

# ============================================================
# 5. MENYIMPAN DAN MENAMPILKAN GRAFIK
# ============================================================

file_gambar = OUTPUT_DIR / "time_series_sinyal_sintetis.png"

fig.savefig(
    file_gambar,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("\nGrafik berhasil dibuat.")
print(f"Lokasi grafik: {file_gambar.resolve()}")

# ============================================================
# 6. HASIL DAN PEMBAHASAN
# ============================================================

print("\n" + "=" * 55)
print("HASIL DAN PEMBAHASAN")
print("=" * 55)

print("""
Grafik menunjukkan perubahan amplitudo sinyal seismik
sintetis terhadap waktu selama 10 sekon.

Sebelum 3 sekon, sinyal didominasi oleh noise acak dengan
amplitudo relatif kecil. Pada waktu 3 sekon, gelombang
sintetis mulai muncul. Setelah itu, amplitudo gelombang
berangsur menurun karena adanya faktor redaman.

Garis vertikal putus-putus menunjukkan waktu impuls.
Label sumbu, legenda, dan grid membantu memperjelas
interpretasi grafik.

Grafik disimpan dalam format PNG dengan resolusi 300 dpi.

Data yang digunakan merupakan data sintetis untuk
praktik pemrograman, bukan rekaman gempa bumi nyata.
""")

# ============================================================
# 7. KESIMPULAN
# ============================================================

print("=" * 55)
print("KESIMPULAN")
print("=" * 55)

print("""
Recipe R-VIS-001 berhasil diterapkan menggunakan Python,
NumPy, dan Matplotlib untuk membuat visualisasi deret waktu.

Grafik memperlihatkan waktu kemunculan gelombang, perubahan
amplitudo, dan pelemahan sinyal akibat redaman dalam simulasi.

Visualisasi ini dapat menjadi contoh dasar pengolahan dan
penyajian data geofisika secara informatif dan reproducible.
""")

# ============================================================
# 8. REFERENSI
# ============================================================

print("=" * 55)
print("REFERENSI")
print("=" * 55)

print("""
AI Cookbook untuk Earth Sciences.
Recipe R-VIS-001: Membuat Time-Series Figure yang Informatif.

Catatan:
Kode ini merupakan implementasi praktik berdasarkan tujuan
recipe, bukan salinan persis kode sumber buku.
""")

print("Proses selesai.")
print(f"Folder hasil: {OUTPUT_DIR.resolve()}")
```
