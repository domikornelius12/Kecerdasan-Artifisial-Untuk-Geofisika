**# 1. STUDI KASUS GEOTHERMAL: DELINEASI REBOISASI DAN RESERVOIR DENGAN MAGNETOTELLURIK (MT)**

* **Role**
  `Anda berperan sebagai seorang Geothermal Geophysicist Senior yang memiliki keahlian mendalam dalam pemodelan data Magnetotellurik (MT) 2D/3D, analisis fasa dan impedansi sistem panas bumi, serta integrasi hidrotermal pada sistem vulkanik aktif.`

* **Task**
  `Lakukan analisis dan interpretasi komprehensif terhadap penampang resistivitas 2D hasil inversi Magnetotellurik (MT) untuk mendelineasi batas-batas sistem panas bumi. Analisis harus mencakup identifikasi ketebalan dan sebaran lapisan penudung (clay cap), batas atas dan samping dari zona reservoir, perkiraan kedalaman sumber panas (heat source), serta penentuan area permeabel yang berpotensi memiliki konektivitas fluida tinggi.`

* **Context**
  `Evaluasi ini dilakukan pada area lapangan panas bumi prospektif di lingkungan gunung api kuarter yang sedang direncanakan untuk tahap pemboran eksplorasi. Pemerintah dan pihak pengembang membutuhkan kepastian mengenai struktur konseptual bawah permukaan guna meminimalkan risiko kegagalan pemboran (dry hole) serta menentukan kedalaman target sumur yang paling optimal.`

* **Input**
  `Data penampang 2D resistivitas hasil inversi MT (frekuensi 0.001 Hz - 1000 Hz), rentang nilai resistivitas konduktif (< 10 ohm.m), resistivitas sedang (30 - 100 ohm.m), dan resistivitas tinggi (> 200 ohm.m), peta geologi permukaan lokal yang menunjukkan zona ubahan hidrotermal (argilik dan propilitik), data manifes permukaan (fumarol dan mata air panas asam/netral), serta data penginderaan jauh terkait struktur kelurusan (lineament).`

* **Constraints**
  `Dalam melakukan interpretasi, jangan secara langsung menyimpulkan bahwa zona dengan resistivitas paling konduktif (< 10 ohm.m) merupakan zona reservoir panas bumi, melainkan harus diinterpretasikan sebagai lapisan penudung (clay cap) berkarakteristik keterubahan mineral smektit. Interpretasi zona reservoir harus dikaitkan dengan zona resistivitas sedang di bawah clay cap yang berasosiasi dengan mineral ubahan temperatur lebih tinggi (illit/klorit). Hindari memberikan estimasi temperatur absolut jika tidak didukung oleh data geotermometer fluida.`

* **Output**
  `Rencana output berupa narasi laporan evaluasi geofisika sistem panas bumi yang disusun secara sistematis, meliputi:`
    `1. Analisis karakteristik tahanan jenis dan delineasi ketebalan serta variasi kedalaman zona clay cap.`
    `2. Pemodelan konseptual geometri reservoir mencakup estimasi kedalaman top of reservoir dan batas lateralnya.`
    `3. Diskusi mengenai hubungan antara kontrol struktur geologi (patahan/rekahan) dengan zona permeabilitas tempat fluida naik.`
    `4. Evaluasi faktor ketidakpastian inversi data MT pada kedalaman dangkal dan sangat dalam.`
    `5. Rekomendasi lokasi dan target kedalaman pemboran sumur landaian suhu atau sumur eksplorasi.`

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LogNorm
from matplotlib.patches import Rectangle

np.random.seed(42)

# 1. PARAMETER MODEL

# Grid model
x = np.linspace(0, 10, 121)       # Jarak lintasan (km)
z = np.linspace(0, 5, 101)        # Kedalaman (km)
X, Z = np.meshgrid(x, z)

# Frekuensi MT
freqs = np.logspace(-3, 3, 80)    # 0.001 - 1000 Hz

# Resistivitas (ohm.m)
rho_clay = 5.0
rho_reservoir = 60.0
rho_bedrock = 300.0
rho_heat_source = 500.0
rho_fault = 15.0

# Geometri hipotetis
clay_top = 0.05                    # km
clay_base = (
    0.65
    + 0.18 * np.sin(x * 0.75)
    + 0.15 * np.exp(-((x - 5.8) / 1.8) ** 2)
)

reservoir_top = 0.80              # km
reservoir_base = 2.50             # km

heat_source_top = 3.00            # km
heat_source_base = 4.50           # km

# Posisi struktur hipotetis
fault_positions = [3.0, 6.5]

# 2. MEMBANGUN MODEL RESISTIVITAS 2D SINTETIS

rho = np.full_like(X, rho_bedrock, dtype=float)

# Lapisan konduktif clay cap
for j, xx in enumerate(x):
    cap = (
        (z >= clay_top)
        & (z <= clay_base[j])
    )
    rho[cap, j] = rho_clay + 1.5 * np.cos(xx)

# Kandidat reservoir resistivitas sedang
reservoir_mask = (
    (Z >= 0.80)
    & (Z <= 2.50)
    & (X >= 1.0)
    & (X <= 8.5)
)

rho[reservoir_mask] = rho_reservoir

# Variasi lateral reservoir
reservoir_lens = (
    ((X - 5.0) / 2.5) ** 2
    + ((Z - 1.55) / 0.85) ** 2
) < 1

rho[reservoir_lens] = 80.0

# Zona sumber panas hipotetis
heat_mask = (
    ((X - 5.2) / 2.0) ** 2
    + ((Z - 3.8) / 1.0) ** 2
) < 1

rho[heat_mask] = rho_heat_source

# Struktur patahan sebagai zona resistivitas relatif rendah
for xf in fault_positions:
    fault = (
        (np.abs(X - xf) < 0.12)
        & (Z > 0.45)
        & (Z < 3.0)
    )
    rho[fault] = rho_fault

# Tambahkan variasi kecil agar model tidak terlalu seragam
rho *= np.exp(
    np.random.normal(0, 0.035, size=rho.shape)
)

rho = np.clip(rho, 1, 1000)

print("=" * 65)
print("MODEL GEOTHERMAL MT 2D SINTETIS")
print("=" * 65)
print(f"Jumlah titik horizontal : {len(x)}")
print(f"Jumlah titik kedalaman : {len(z)}")
print(f"Rentang frekuensi MT   : {freqs.min():.3f} - {freqs.max():.0f} Hz")
print(f"Resistivitas minimum   : {rho.min():.2f} ohm.m")
print(f"Resistivitas maksimum  : {rho.max():.2f} ohm.m")
print("Model berhasil dibuat.")


# 3. ANALISIS CLAY CAP DAN RESERVOIR

# Ketebalan clay cap di setiap posisi horizontal
clay_thickness = clay_base - clay_top

print("\n" + "=" * 65)
print("HASIL DELINEASI CLAY CAP")
print("=" * 65)

print(f"Kedalaman atas clay cap  : {clay_top:.2f} km")
print(f"Kedalaman dasar minimum  : {clay_base.min():.2f} km")
print(f"Kedalaman dasar maksimum : {clay_base.max():.2f} km")
print(f"Ketebalan minimum        : {clay_thickness.min():.2f} km")
print(f"Ketebalan maksimum       : {clay_thickness.max():.2f} km")
print(f"Ketebalan rata-rata      : {clay_thickness.mean():.2f} km")

print("\n" + "=" * 65)
print("GEOMETRI KANDIDAT RESERVOIR")
print("=" * 65)

print(f"Top reservoir             : {reservoir_top:.2f} km")
print(f"Base reservoir konseptual : {reservoir_base:.2f} km")
print(f"Ketebalan interval        : {reservoir_base-reservoir_top:.2f} km")
print("Sebaran horizontal        : 1.0 - 8.5 km")
print("Resistivitas representatif: 30 - 100 ohm.m")


# 4. ESTIMASI SKIN DEPTH MT

# Skin depth pendekatan untuk medium homogen:
# delta (m) = 503 * sqrt(rho / f)
#
# Ini bukan kedalaman investigasi aktual.
# Pada bumi berlapis, respons MT bergantung pada seluruh
# distribusi resistivitas dan frekuensi.

rho_reference = 100.0
skin_depth = 503 * np.sqrt(rho_reference / freqs)

print("\n" + "=" * 65)
print("ESTIMASI SKIN DEPTH MT")
print("=" * 65)

for f in [0.001, 0.01, 0.1, 1, 10, 100, 1000]:
    delta = 503 * np.sqrt(rho_reference / f)
    print(f"Frekuensi {f:8.3f} Hz -> Skin depth ~ {delta/1000:.2f} km")


# 5. MODEL SINTETIS RESPONS RESISTIVITAS SEMU

# Pendekatan ilustratif untuk membangun kurva frekuensi.
# BUKAN forward modeling MT Maxwell 2D.
#
# Respons berikut hanya menunjukkan tren hipotetis,
# bukan respons observasi atau hasil inversi.

rho_app = np.zeros_like(freqs)

for i, f in enumerate(freqs):
    delta_km = (
        503 * np.sqrt(rho_reference / f)
    ) / 1000

    if delta_km < 0.7:
        rho_app[i] = 8 + 3 * np.log10(1 + delta_km)
    elif delta_km < 2.5:
        rho_app[i] = 45 + 15 * np.log10(delta_km + 1)
    else:
        rho_app[i] = 180 + 45 * np.log10(delta_km)

# Noise ilustratif
rho_app_obs = rho_app * np.exp(
    np.random.normal(0, 0.08, size=len(freqs))
)

print("\nKurva resistivitas semu sintetis berhasil dibuat.")
print("Kurva tidak mewakili data MT lapangan.")


# 6. ANALISIS STRUKTUR DAN KANDIDAT PERMEABILITAS

print("\n" + "=" * 65)
print("ANALISIS STRUKTUR GEOLOGI")
print("=" * 65)

for i, xf in enumerate(fault_positions, start=1):
    print(f"Struktur hipotetis {i}: x = {xf:.2f} km")

print("""
Interpretasi konseptual:
- Patahan dapat menjadi jalur aliran fluida atau penghalang.
- Perpotongan patahan dengan kandidat reservoir perlu dievaluasi.
- Manifestasi fumarol dan mata air panas dapat mendukung model.
- MT tidak mengukur permeabilitas secara langsung.
- Konfirmasi memerlukan data geologi, geokimia, dan data sumur.
""")


# 7. VISUALISASI HASIL MODEL MT

fig, axes = plt.subplots(1, 3, figsize=(19, 7))

# ------------------------------------------------------------
# PANEL 1: PENAMPANG RESISTIVITAS 2D
# ------------------------------------------------------------

ax = axes[0]

im = ax.pcolormesh(
    X, Z, rho,
    shading="auto",
    cmap="turbo",
    norm=LogNorm(vmin=1, vmax=1000)
)

cb = plt.colorbar(im, ax=ax, pad=0.02)
cb.set_label("Resistivitas (ohm.m)")

# Garis batas clay cap
ax.plot(
    x, clay_base,
    color="white",
    linestyle="--",
    linewidth=2,
    label="Dasar clay cap"
)

# Batas reservoir konseptual
ax.axhline(
    reservoir_top,
    color="black",
    linestyle=":",
    linewidth=1.8,
    label="Top reservoir konseptual"
)

ax.axhline(
    reservoir_base,
    color="black",
    linestyle="-.",
    linewidth=1.8,
    label="Base reservoir konseptual"
)

# Posisi struktur
for xf in fault_positions:
    ax.axvline(
        xf,
        color="white",
        linestyle=":",
        linewidth=1.4
    )

# Sumber panas
ax.contour(
    X, Z, heat_mask.astype(float),
    levels=[0.5],
    colors="black",
    linewidths=1.5
)

ax.set_ylim(5, 0)
ax.set_xlabel("Jarak lintasan (km)")
ax.set_ylabel("Kedalaman (km)")
ax.set_title("Model Resistivitas MT 2D", fontweight="bold")
ax.legend(fontsize=8, loc="lower right")


# ------------------------------------------------------------
# PANEL 2: KURVA RESISTIVITAS SEMU SINTETIS
# ------------------------------------------------------------

ax = axes[1]

ax.semilogx(
    freqs,
    rho_app_obs,
    "bo",
    markersize=4,
    alpha=0.7,
    label="Respons sintetis + noise"
)

ax.semilogx(
    freqs,
    rho_app,
    "r-",
    linewidth=2,
    label="Respons ilustratif"
)

ax.set_xlabel("Frekuensi (Hz)")
ax.set_ylabel("Resistivitas semu (ohm.m)")
ax.set_title("Kurva Resistivitas Semu", fontweight="bold")
ax.grid(True, which="both", alpha=0.25)
ax.legend(fontsize=8)


# ------------------------------------------------------------
# PANEL 3: PROFIL KETEBALAN CLAY CAP
# ------------------------------------------------------------

ax = axes[2]

ax.plot(
    x,
    clay_thickness * 1000,
    color="purple",
    linewidth=2.5,
    label="Ketebalan clay cap"
)

ax.fill_between(
    x,
    0,
    clay_thickness * 1000,
    alpha=0.2,
    color="purple"
)

for xf in fault_positions:
    ax.axvline(
        xf,
        color="red",
        linestyle="--",
        alpha=0.7,
        label="Struktur hipotetis" if xf == fault_positions[0] else None
    )

ax.set_xlabel("Jarak lintasan (km)")
ax.set_ylabel("Ketebalan (m)")
ax.set_title("Variasi Ketebalan Clay Cap", fontweight="bold")
ax.grid(True, alpha=0.25)
ax.legend(fontsize=8)


plt.suptitle(
    "ANALISIS GEOTHERMAL MAGNETOTELLURIK (MT) - MODEL SINTETIS",
    fontsize=14,
    fontweight="bold"
)

plt.tight_layout()
plt.savefig(
    "geothermal_mt_synthetic.png",
    dpi=200,
    bbox_inches="tight"
)
plt.show()

print("\nVisualisasi berhasil dibuat.")
print("File gambar: geothermal_mt_synthetic.png")


# 8. REKOMENDASI TARGET EKSPLORASI

print("\n" + "=" * 65)
print("REKOMENDASI TARGET EKSPLORASI")
print("=" * 65)

print("""
1. TARGET SURVEI LANJUTAN
   Periksa zona perubahan geometri clay cap dan hubungan
   struktur dengan kandidat reservoir.

2. TARGET SUMUR LANDAIAN SUHU
   Tentukan setelah evaluasi geologi, akses, risiko lingkungan,
   dan tujuan pengujian gradien temperatur.

3. TARGET SUMUR EKSPLORASI
   Prioritaskan area yang didukung bersama oleh:
   - Kandidat reservoir di bawah clay cap.
   - Struktur yang berpotensi mengontrol sirkulasi fluida.
   - Alterasi hidrotermal dan manifestasi permukaan.
   - Hasil pemodelan dan analisis ketidakpastian.

4. KEDALAMAN PEMBORAN
   Kedalaman target harus ditentukan dari data aktual dan
   kajian terpadu. Nilai kedalaman dalam model ini hanya
   parameter ilustratif.

5. VALIDASI
   Gunakan kurva resistivitas semu dan fase hasil pengukuran,
   quality control, forward modeling, serta inversi MT yang
   sesuai sebelum membuat keputusan pemboran.
""")


# 9. KESIMPULAN

print("\n" + "=" * 65)
print("KESIMPULAN")
print("=" * 65)

print(f"""
Model sintetis menunjukkan kandidat clay cap dengan
ketebalan rata-rata sekitar {clay_thickness.mean():.2f} km.

Kandidat reservoir konseptual ditempatkan pada interval
{reservoir_top:.2f} - {reservoir_base:.2f} km.

Sumber panas hipotetis dimodelkan pada interval
{heat_source_top:.2f} - {heat_source_base:.2f} km.

Struktur pada posisi lintasan {fault_positions} km digunakan
sebagai contoh jalur yang perlu diuji keterkaitannya dengan
zona reservoir dan manifestasi hidrotermal.

Semua kedalaman dan respons listrik dalam demonstrasi ini
bersifat sintetis. Model bukan hasil inversi MT aktual,
dan tidak boleh digunakan langsung untuk menentukan lokasi
atau kedalaman pemboran.
""")
```

**# 2. STUDI KASUS PERTAMBANGAN: EKSPLORASI BIJIH BESI DENGAN METODE GEOMAGNETIK**

* **Role**
  `Anda berperan sebagai seorang Mining Geophysicist ahli yang berpengalaman dalam akuisisi, pengolahan data medan magnetik bumi, aplikasi pemrosesan sinyal (filter penataan/transformasi medan), serta pemodelan geometri tubuh bijih kaya mineral magnetit.`

* **Task**
  `Lakukan evaluasi dan interpretasi menyeluruh terhadap data anomali magnetik total beserta hasil filter pengolahan citra guna memetakan sebaran lateral, kedalaman puncak, dan estimasi kemiringan (dip) dari tubuh bijih besi (magnetit). Anda diminta untuk memisahkan respon anomali magnetik yang bersumber dari mineralisasi utama dengan respon anomali latar belakang akibat batuan samping.`

* **Context**
  `Survei magnetik darat dilakukan pada suatu wilayah prospek pertambangan bijih besi yang sebagian besar permukaannya tertutup oleh endapan aluvial dan tanah pelapukan (overburden) yang tebal. Informasi mengenai batas bawah permukaan dari zona mineralisasi sangat dibutuhkan oleh tim geologi tambang untuk merencanakan titik-titik pemboran inti (core drilling) secara efisien.`

* **Input**
  `Peta Anomali Magnetik Total (Total Magnetic Anomaly - TMA), peta hasil transformasi Reduce to Pole (RTP), peta Analytic Signal, peta First Vertical Derivative (1VD), data pengukuran suseptibilitas magnetik dari sampel batuan inti/singkapan lokal, serta data ketebalan tanah penutup berdasarkan beberapa data parit uji (trenching).`

* **Constraints**
  `Analisis harus memperhitungkan efek dari kemagnetan sisa (remanent magnetization) yang berpotensi menyebabkan pergeseran posisi anomali puncak pada peta RTP. Jangan mengasumsikan seluruh nilai anomali magnetik tinggi sebagai tubuh bijih murni tanpa mempertimbangkan variasi suseptibilitas batuan samping (seperti batuan basal atau andesit). Wajib menjelaskan ambiguitas pemodelan magnetik yang bersifat non-unique.`

* **Output**
  `Rencana output berupa narasi ilmiah-teknis yang terstruktur, memuat:`
  `1. Penjelasan prinsip fisik respon magnetik terhadap kandungan mineral magnetit.`
  `2. Analisis perbandingan peta TMA, RTP, dan Analytic Signal dalam menentukan pusat kemagnetan tubuh bijih.`
  `3. Interpretasi struktur geologi pengontrol mineralisasi (seperti zona sesar atau kontak batuan) berdasarkan peta derivatif (1VD).`
  `4. Perkiraan dimensi geometri, kedalaman puncak (top depth), dan arah kemiringan tubuh mineralisasi.`
  `5. Penilaian tingkat ketidakpastian data dan rekomendasi koordinat titik bor verifikasi.`

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import TwoSlopeNorm

np.random.seed(42)

# 1. PARAMETER SURVEI DAN MODEL GEOLOGI

# Grid survei (meter)
x = np.linspace(0, 1000, 151)
y = np.linspace(0, 800, 121)
X, Y = np.meshgrid(x, y)

dx = x[1] - x[0]
dy = y[1] - y[0]

# Medan magnetik bumi dan asumsi arah magnetisasi
inclination = -30.0       # derajat, asumsi ilustratif
declination = 1.0         # derajat, asumsi ilustratif
remanent_magnetization = True

# Geometri tubuh bijih sintetis
ore_x = 520.0             # pusat tubuh bijih (m)
ore_y = 400.0
ore_top_true = 80.0       # kedalaman puncak tubuh (m)
ore_width = 220.0         # lebar karakteristik (m)
ore_length = 420.0        # panjang karakteristik (m)
ore_dip_true = 65.0       # derajat dari horizontal
ore_strike = 35.0        # orientasi sumbu tubuh (derajat)

# Suseptibilitas volume (SI), nilai ilustratif
k_magnetite = 0.08
k_basalt = 0.025
k_andesite = 0.008
k_host = 0.001

# Overburden
overburden_mean = 25.0
overburden_max = 60.0

# Anomali sintetis, nT
ore_amplitude = 450.0
basalt_amplitude = 170.0
regional_gradient = 0.08

print("=" * 65)
print("SIMULASI EKSPLORASI BIJIH BESI - GEOMAGNETIK")
print("=" * 65)
print(f"Ukuran grid       : {X.shape[1]} x {X.shape[0]} titik")
print(f"Spasi horizontal  : {dx:.1f} m")
print(f"Spasi vertikal   : {dy:.1f} m")
print(f"Suseptibilitas magnetit : {k_magnetite:.3f} SI")
print(f"Suseptibilitas basalt   : {k_basalt:.3f} SI")
print(f"Top tubuh sintetis      : {ore_top_true:.1f} m")
print(f"Dip tubuh sintetis      : {ore_dip_true:.1f} derajat")


# 2. MEMBANGUN ANOMALI MAGNETIK TOTAL (TMA) SINTETIS

def rotated_gaussian(X, Y, xc, yc, sx, sy, angle_deg):
    """Membuat anomali elips yang diputar."""
    theta = np.radians(angle_deg)
    xr = (X - xc) * np.cos(theta) + (Y - yc) * np.sin(theta)
    yr = -(X - xc) * np.sin(theta) + (Y - yc) * np.cos(theta)

    return np.exp(
        -0.5 * ((xr / sx) ** 2 + (yr / sy) ** 2)
    )


# Respons tubuh magnetit
ore_response = ore_amplitude * rotated_gaussian(
    X, Y,
    ore_x + 45, ore_y - 35,   # pergeseran karena arah magnetisasi
    ore_width / 2.5,
    ore_length / 2.5,
    ore_strike
)

# Respons batuan samping basalt
basalt_response = basalt_amplitude * rotated_gaussian(
    X, Y, 280, 520, 170, 110, -20
)

# Respons andesit yang lebih lemah
andesite_response = 65 * rotated_gaussian(
    X, Y, 750, 230, 140, 100, 25
)

# Anomali regional
regional = (
    regional_gradient * (X - 500)
    + 0.035 * (Y - 400)
)

# Anomali total sintetis dengan noise
noise = np.random.normal(0, 4.0, X.shape)

TMA = (
    ore_response
    + basalt_response
    + andesite_response
    + regional
    + noise
)

print("\nTMA sintetis berhasil dibangun.")
print(f"Minimum TMA : {TMA.min():.2f} nT")
print(f"Maksimum TMA: {TMA.max():.2f} nT")


# 3. KOREKSI REGIONAL DAN RTP ILUSTRATIF

# Estimasi background regional sederhana.
# Pada data nyata, regional-residual separation perlu diuji
# dengan metode yang sesuai dan tidak selalu berupa bidang datar.

regional_est = (
    regional_gradient * (X - 500)
    + 0.035 * (Y - 400)
)

residual = TMA - regional_est

# RTP ilustratif:
# Pergeseran posisi anomali untuk menunjukkan konsep koreksi.
# BUKAN RTP sebenarnya dan tidak menggunakan operator Fourier.
# Pada data nyata, RTP membutuhkan inklinasi, deklinasi,
# arah magnetisasi, serta evaluasi remanensi.

shift_x = -45.0
shift_y = 35.0

RTP_illustrative = (
    ore_amplitude * rotated_gaussian(
        X, Y, ore_x, ore_y,
        ore_width / 2.5,
        ore_length / 2.5,
        ore_strike
    )
    + (basalt_response + andesite_response)
)

RTP_illustrative += np.random.normal(0, 3.0, X.shape)

print("\nPemisahan regional dan residual selesai.")
print("RTP yang dibuat adalah visualisasi ilustratif, bukan transformasi fisik RTP.")


# 4. ANALYTIC SIGNAL DAN FIRST VERTICAL DERIVATIVE
# Turunan horizontal
dT_dx = np.gradient(residual, dx, axis=1)
dT_dy = np.gradient(residual, dy, axis=0)

# Turunan vertikal semu berbasis spektrum Fourier.
# Asumsi permukaan pengamatan datar dan grid teratur.
# Hasil tetap dipengaruhi noise, ukuran grid, dan tepi peta.

ny, nx = residual.shape
kx = 2 * np.pi * np.fft.fftfreq(nx, d=dx)
ky = 2 * np.pi * np.fft.fftfreq(ny, d=dy)
KX, KY = np.meshgrid(kx, ky)
K = np.sqrt(KX**2 + KY**2)

F = np.fft.fft2(residual)

# 1VD: operator spektral |k|
# Ini adalah pendekatan turunan vertikal pertama dalam
# domain Fourier untuk medan potensial di permukaan.
F_1VD = K * F
VD1 = np.real(np.fft.ifft2(F_1VD))

# Analytic signal 3D untuk data medan potensial:
# AS = sqrt((dT/dx)^2 + (dT/dy)^2 + (dT/dz)^2)
AS = np.sqrt(dT_dx**2 + dT_dy**2 + VD1**2)

print("\nFilter berhasil dihitung:")
print(f"AS maksimum  : {AS.max():.2f}")
print(f"1VD minimum  : {VD1.min():.2f} nT/m")
print(f"1VD maksimum : {VD1.max():.2f} nT/m")


# 5. ESTIMASI PUSAT ANOMALI

# Maksimum analytic signal digunakan sebagai kandidat pusat
# batas sumber magnetik. Nilai ini tidak otomatis sama dengan
# pusat bijih karena batuan samping juga dapat menghasilkan AS.

peak_idx = np.unravel_index(np.argmax(AS), AS.shape)

peak_x = X[peak_idx]
peak_y = Y[peak_idx]
peak_as = AS[peak_idx]

print("\n" + "=" * 65)
print("KANDIDAT PUSAT SUMBER MAGNETIK")
print("=" * 65)
print(f"X kandidat : {peak_x:.1f} m")
print(f"Y kandidat : {peak_y:.1f} m")
print(f"Nilai AS   : {peak_as:.2f}")


# 6. ESTIMASI KEDALAMAN AWAL DARI LEBAR ANOMALI

# Perkiraan setengah lebar anomali sepanjang profil melalui
# kandidat pusat. Metode setengah lebar sangat sederhana dan
# hanya digunakan sebagai screening awal untuk sumber kompak.
#
# Ini bukan inversi geometri, dan tidak berlaku universal
# untuk tubuh berbentuk lembaran, remanensi kuat, atau
# beberapa sumber yang saling tumpang tindih.

profile = residual[
    np.argmin(np.abs(y - peak_y)), :
]

profile_abs = np.abs(profile)
peak_profile_idx = np.argmax(profile_abs)
peak_profile = profile_abs[peak_profile_idx]

half_level = peak_profile / 2.0
left_candidates = np.where(
    profile_abs[:peak_profile_idx] <= half_level
)[0]
right_candidates = np.where(
    profile_abs[peak_profile_idx + 1:] <= half_level
)[0]

if len(left_candidates) > 0 and len(right_candidates) > 0:
    i_left = left_candidates[-1]
    i_right = peak_profile_idx + 1 + right_candidates[0]

    half_width = abs(x[i_right] - x[i_left]) / 2.0

    # Aturan skala ilustratif, bukan formula kedalaman universal.
    depth_screening = half_width * 0.5
else:
    half_width = np.nan
    depth_screening = np.nan

print("\n" + "=" * 65)
print("SCREENING KEDALAMAN SUMBER")
print("=" * 65)

if np.isfinite(depth_screening):
    print(f"Setengah lebar anomali : {half_width:.1f} m")
    print(f"Perkiraan awal depth   : {depth_screening:.1f} m")
    print("Catatan: nilai ini hanya screening awal.")
else:
    print("Lebar anomali tidak berhasil ditentukan secara stabil.")

print(f"Top tubuh pada model sintetis: {ore_top_true:.1f} m")
print("Top model sintetis diketahui dari parameter pembangkitan,")
print("bukan hasil inversi data magnetik.")


# 7. ESTIMASI ORIENTASI DAN DIP

# Analisis orientasi elips dari koordinat berbobot analytic signal.
# Orientasi merupakan arah sumbu dominan anomali, bukan otomatis
# strike tubuh bijih yang sesungguhnya.

weights = np.clip(AS - np.percentile(AS, 70), 0, None)

if weights.sum() > 0:
    x_mean = np.sum(X * weights) / weights.sum()
    y_mean = np.sum(Y * weights) / weights.sum()

    cov_xx = np.sum(weights * (X - x_mean) ** 2) / weights.sum()
    cov_yy = np.sum(weights * (Y - y_mean) ** 2) / weights.sum()
    cov_xy = np.sum(weights * (X - x_mean) * (Y - y_mean)) / weights.sum()

    covariance = np.array([
        [cov_xx, cov_xy],
        [cov_xy, cov_yy]
    ])

    eigvals, eigvecs = np.linalg.eigh(covariance)
    major_axis = eigvecs[:, np.argmax(eigvals)]

    strike_est = np.degrees(
        np.arctan2(major_axis[1], major_axis[0])
    )

    print("\nOrientasi dominan anomali (ilustratif):")
    print(f"Strike hasil PCA: {strike_est:.1f} derajat")
else:
    strike_est = np.nan
    print("\nOrientasi tidak dapat ditentukan.")

print(f"Dip model sintetis sebenarnya: {ore_dip_true:.1f} derajat")
print("Dip tidak dapat ditentukan secara unik dari peta anomali")
print("saja. Diperlukan pemodelan profil, informasi geologi,")
print("dan pengujian beberapa geometri alternatif.")


# 8. INTERPRETASI SUSEPTIBILITAS

print("\n" + "=" * 65)
print("INTERPRETASI SUSEPTIBILITAS MAGNETIK")
print("=" * 65)

print(f"Magnetit   : {k_magnetite:.3f} SI")
print(f"Basalt     : {k_basalt:.3f} SI")
print(f"Andesit    : {k_andesite:.3f} SI")
print(f"Batuan host: {k_host:.3f} SI")

print("""
Interpretasi:
- Magnetit dapat menghasilkan respons magnetik yang kuat.
- Basalt juga dapat memiliki suseptibilitas tinggi dan
  menjadi sumber anomali yang menyerupai mineralisasi.
- Andesit dan batuan samping lainnya memiliki rentang
  suseptibilitas yang bervariasi.
- Nilai suseptibilitas tidak langsung menunjukkan kadar bijih.
- Remanensi dapat mengubah arah dan posisi respons magnetik.
- Verifikasi membutuhkan sampel, petrologi, dan data geologi.
""")


# 9. VISUALISASI 4 PANEL

fig, axes = plt.subplots(2, 2, figsize=(16, 12))

panels = [
    (axes[0, 0], TMA, "Total Magnetic Anomaly (TMA)", "nT"),
    (axes[0, 1], RTP_illustrative, "RTP Ilustratif (bukan transformasi RTP)", "nT"),
    (axes[1, 0], AS, "Analytic Signal", "nT/m"),
    (axes[1, 1], VD1, "First Vertical Derivative (1VD)", "nT/m"),
]

for ax, data, title, unit in panels:
    vmax = np.percentile(np.abs(data), 98)

    if np.min(data) < 0 < np.max(data):
        norm = TwoSlopeNorm(vmin=-vmax, vcenter=0, vmax=vmax)
        im = ax.pcolormesh(
            X, Y, data,
            shading="auto",
            cmap="RdBu_r",
            norm=norm
        )
    else:
        im = ax.pcolormesh(
            X, Y, data,
            shading="auto",
            cmap="viridis"
        )

    plt.colorbar(im, ax=ax, label=unit)
    ax.set_xlabel("X (m)")
    ax.set_ylabel("Y (m)")
    ax.set_title(title, fontweight="bold")

    # Kandidat pusat AS
    ax.plot(
        peak_x, peak_y,
        marker="x",
        color="black",
        markersize=10,
        markeredgewidth=2,
        label="Kandidat pusat AS"
    )

    # Posisi pusat model sintetis
    ax.plot(
        ore_x, ore_y,
        marker="+",
        color="yellow",
        markersize=10,
        markeredgewidth=2,
        label="Pusat model bijih"
    )

    ax.legend(fontsize=8, loc="upper right")

plt.suptitle(
    "EKSPLORASI BIJIH BESI DENGAN GEOMAGNETIK - MODEL SINTETIS",
    fontsize=14,
    fontweight="bold"
)

plt.tight_layout()
plt.savefig(
    "geomagnetik_bijih_besi.png",
    dpi=200,
    bbox_inches="tight"
)
plt.show()

print("\nGambar tersimpan: geomagnetik_bijih_besi.png")


# 10. PROFIL ANOMALI DAN SKEMA GEOMETRI

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Profil residual
ax = axes[0]
iy = np.argmin(np.abs(y - peak_y))

ax.plot(
    x, residual[iy, :],
    color="navy",
    linewidth=2,
    label="Residual magnetik"
)

ax.axvline(
    ore_x,
    color="red",
    linestyle="--",
    label="Pusat model bijih"
)

ax.axvline(
    peak_x,
    color="green",
    linestyle=":",
    label="Kandidat pusat AS"
)

ax.set_xlabel("Jarak X (m)")
ax.set_ylabel("Anomali residual (nT)")
ax.set_title("Profil Anomali Magnetik")
ax.grid(True, alpha=0.3)
ax.legend()

# Skema konseptual bawah permukaan
ax = axes[1]

ax.set_xlim(0, 1000)
ax.set_ylim(500, 0)

# Overburden
ax.add_patch(Rectangle(
    (0, 0), 1000, overburden_mean,
    facecolor="tan", alpha=0.7,
    label="Overburden (skema)"
))

# Tubuh bijih sebagai lembaran miring
ore_polygon_x = [400, 480, 650, 570]
ore_polygon_z = [380, 90, 150, 440]

ax.fill(
    ore_polygon_x,
    ore_polygon_z,
    color="darkred",
    alpha=0.7,
    label="Tubuh bijih konseptual"
)

# Titik kandidat bor berdasarkan pusat model.
# Ini hanya titik verifikasi sintetis, bukan koordinat lapangan.
drill_points = [
    (ore_x, "B1"),
    (ore_x + 100, "B2"),
    (ore_x - 100, "B3")
]

for xd, label in drill_points:
    ax.plot(
        xd, overburden_mean,
        marker="v",
        color="blue",
        markersize=9
    )
    ax.text(
        xd + 8, overburden_mean + 18,
        label,
        fontsize=9
    )

ax.set_xlabel("Jarak lintasan (m)")
ax.set_ylabel("Kedalaman (m)")
ax.set_title("Skema Konseptual dan Titik Bor Verifikasi")
ax.grid(True, alpha=0.2)
ax.legend(fontsize=8, loc="lower right")

plt.tight_layout()
plt.savefig(
    "skema_target_bor.png",
    dpi=200,
    bbox_inches="tight"
)
plt.show()

# 11. REKOMENDASI TITIK BOR VERIFIKASI

print("\n" + "=" * 65)
print("REKOMENDASI TITIK BOR VERIFIKASI")
print("=" * 65)

print("""
Titik bor berikut merupakan contoh posisi relatif terhadap
model sintetis. Bukan koordinat untuk pelaksanaan pemboran.

B1: Sekitar pusat tubuh bijih model.
    Tujuan: memverifikasi litologi, mineralisasi, dan kedalaman top.

B2: Sekitar 100 m di sisi timur pusat model.
    Tujuan: menguji kontinuitas lateral tubuh bijih.

B3: Sekitar 100 m di sisi barat pusat model.
    Tujuan: menguji batas lateral dan variasi batuan samping.

Penentuan koordinat UTM aktual harus dilakukan dengan
mengintegrasikan peta geologi, hasil pemodelan magnetik,
ketebalan overburden, akses pemboran, serta batas lahan.
""")

for xd, label in drill_points:
    print(f"{label}: X lokal = {xd:.1f} m, Y lokal = {overburden_mean:.1f} m (skema permukaan)")

# 12. KESIMPULAN

print("\n" + "=" * 65)
print("KESIMPULAN EVALUASI GEOMAGNETIK")
print("=" * 65)

print(f"""
1. TMA memperlihatkan gabungan respons tubuh bijih dan
   batuan samping yang memiliki sifat magnetik.

2. Peta RTP idealnya membantu memusatkan anomali pada sumber
   jika arah magnetisasi diketahui dan asumsi transformasi
   terpenuhi. Remanensi yang tidak diketahui dapat menggeser
   atau mengubah bentuk anomali hasil RTP.

3. Analytic Signal membantu menandai batas lateral sumber
   magnetik, tetapi tidak secara langsung membuktikan kadar bijih.

4. First Vertical Derivative menonjolkan perubahan medan
   berfrekuensi spasial tinggi dan dapat membantu memetakan
   kontak litologi serta struktur.

5. Top tubuh bijih, dimensi, dan dip perlu diuji melalui
   pemodelan maju dan inversi dengan beberapa geometri alternatif.

6. Pemboran verifikasi diprioritaskan pada target yang konsisten
   dengan geologi, suseptibilitas, hasil filter, dan data
   ketebalan overburden.

Model ini sintetis sehingga tidak memberikan estimasi cadangan
atau kepastian mineralisasi aktual.
""")
```

**# 3. STUDI KASUS MITIGASI BENCANA: PEMETAAN ZONA RENTAN LIKUIFAKSI MENGGUNAKAN METODE SEISMIK REFRAKSI**

* **Role**
  `Anda berperan sebagai seorang Geophysicist Bencana dan Geoteknik yang memiliki spesialisasi dalam bidang analisis dinamika tanah, pemrosesan sinyal seismik refraksi, tomografi gelombang kompresi (Vp), serta penilaian risiko kebencanaan akibat guncangan gempa bumi.`

* **Task**
  `Menganalisis profil kecepatan gelombang kompresi 2D (Vp) hasil tomografi refraksi seismik untuk memetakan variasi ketebalan sedimen lunak, kedalaman lapisan batuan dasar (bedrock), serta mengidentifikasi zona perlapisan pasir lepas jenuh air yang memiliki kerentanan tinggi terhadap fenomena likuifaksi.`

* **Context**
  `Investigasi geofisika dangkal ini dilakukan di kawasan pemukiman padat penduduk yang berada dekat dengan zona sesar aktif. Wilayah ini berdiri di atas endapan aluvial dan pernah mengalami guncangan gempa. Pemerintah daerah memerlukan peta zonasi kerentanan tanah sebagai dasar penyusunan tata ruang dan standar keteknikan fondasi bangunan baru.`

* **Input**
  `Data waktu tempuh gelombang pertama (first-break travel time), penampang profil kecepatan Vp 2D hasil inversi tomografi, data variasi kedalaman muka air tanah dangkal hasil pengukuran sumur warga (1 - 4 meter), deskripsi sampel pemboran geoteknik dangkal, serta data percepatan batuan dasar rencana.`

* **Constraints**
  `Dalam menentukan potensi likuifaksi, jangan hanya mengandalkan satu nilai Vp semata, tetapi wajib mengintegrasikan analisis antara nilai Vp rendah hingga sedang (300 m/s - 1200 m/s), tingkat kejenuhan air (di mana Vp pada sedimen jenuh air umumnya melonjak mendekati 1500 m/s akibat sifat fluida), dan keberadaan muka air tanah dangkal. Jelaskan secara tegas keterbatasan seismik refraksi dalam membedakan batas antar-lapisan jika terjadi kondisi Vp yang mengalami velocity reversal (lapisan lambat di bawah lapisan cepat).`

* **Output**
  `Rencana output berupa narasi laporan komprehensif mitigasi bencana, meliputi:`
  `1. Interpretasi pola distribusi kecepatan gelombang Vp terhadap kedalaman dan jarak lintasan.`
  `2. Pemetaan ketebalan lapisan sedimen permukaan dan delineasi profil kedalaman bedrock.`
  `3. Analisis tumpang-tindih (overlay) antara zona Vp jenuh air dengan kriteria material pasir lepas rawan likuifaksi.`
  `4. Diskusi mengenai faktor ketidakpastian tomografi refraksi dan batasan resolusi lateral.`
  `5. Kesimpulan kondisi tapak (site condition) serta rekomendasi pengujian geoteknik lanjutan (seperti CPT atau SPT).`

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
from matplotlib.patches import Patch

np.random.seed(42)

# 1. PARAMETER SURVEI

panjang_lintasan = 300       # meter
kedalaman_maksimum = 40      # meter
jarak_grid = 2               # meter

x = np.arange(0, panjang_lintasan + jarak_grid, jarak_grid)
z = np.arange(0, kedalaman_maksimum + jarak_grid, jarak_grid)

X, Z = np.meshgrid(x, z)

print("=" * 65)
print("SIMULASI TOMOGRAFI SEISMIK REFRAKSI UNTUK LIKUIFAKSI")
print("=" * 65)
print(f"Panjang lintasan       : {panjang_lintasan} m")
print(f"Kedalaman investigasi  : {kedalaman_maksimum} m")
print(f"Spasi grid             : {jarak_grid} m")
print(f"Jumlah titik grid      : {X.size}")


# 2. MEMBANGUN MODEL GEOLOGI SINTETIS
# Kode ini membangun model konseptual:
# - Lapisan tanah permukaan
# - Sedimen aluvial
# - Lensa pasir lepas
# - Lapisan lebih kompak
# - Bedrock
#
# Semua nilai dan geometri merupakan asumsi simulasi.

# Kedalaman batas bedrock dibuat bervariasi secara lateral
kedalaman_bedrock = (
    25
    + 3.0 * np.sin(2 * np.pi * x / 180)
    + 1.5 * np.cos(2 * np.pi * x / 95)
)

bedrock_grid = np.tile(kedalaman_bedrock, (len(z), 1))

# Model kecepatan awal
Vp = np.zeros_like(X, dtype=float)

# Lapisan permukaan: tanah lepas
Vp[Z < 3] = 350

# Sedimen aluvial: lempung lanauan dan material campuran
mask_aluvial = (Z >= 3) & (Z < 10)
Vp[mask_aluvial] = 650

# Sedimen lebih padat
mask_kompak = (Z >= 10) & (Z < bedrock_grid)
Vp[mask_kompak] = 1150

# Bedrock dengan kecepatan lebih tinggi
Vp[Z >= bedrock_grid] = 2600

# Membentuk lensa pasir lepas di beberapa bagian lintasan
# Bentuk ini hanya contoh heterogenitas lateral.
lensa1 = (
    ((X - 75) / 42) ** 2
    + ((Z - 7.0) / 3.0) ** 2
) <= 1

lensa2 = (
    ((X - 205) / 38) ** 2
    + ((Z - 9.0) / 3.5) ** 2
) <= 1

# Pasir lepas relatif kering/tidak jenuh dalam model awal
Vp[lensa1 | lensa2] = 850

# Memodelkan pengaruh kejenuhan air pada sebagian sedimen.
# Vp dapat meningkat akibat pengaruh fluida pori, tetapi
# respons sebenarnya bergantung pada porositas, rangka butir,
# tekanan, komposisi, dan kondisi geologi.
zona_jenuh1 = (
    (X >= 40) & (X <= 115)
    & (Z >= 4) & (Z <= 10)
)

zona_jenuh2 = (
    (X >= 170) & (X <= 245)
    & (Z >= 5) & (Z <= 12)
)

# Bagian pasir yang diasumsikan jenuh air
Vp[zona_jenuh1 & (lensa1)] = 1450
Vp[zona_jenuh2 & (lensa2)] = 1500

# Variasi kecil untuk merepresentasikan heterogenitas
noise = np.random.normal(0, 25, size=Vp.shape)

# Jangan menambahkan noise berlebihan pada bedrock
Vp = Vp + noise
Vp = np.clip(Vp, 250, 3200)

# 3. MODEL MUKA AIR TANAH
# Kedalaman muka air tanah diasumsikan bervariasi 1-4 meter.
# Nilai ini mewakili pengukuran sumur hipotetis.

muka_air = (
    2.5
    + 0.9 * np.sin(2 * np.pi * x / 150)
    + 0.45 * np.cos(2 * np.pi * x / 75)
)

muka_air = np.clip(muka_air, 1.0, 4.0)
muka_air_grid = np.tile(muka_air, (len(z), 1))

# Zona di bawah muka air tanah
mask_di_bawah_air = Z >= muka_air_grid


# 4. MODEL LITOLOGI SINTETIS
# Kode litologi:
# 0 = tanah permukaan / material campuran
# 1 = sedimen halus atau aluvial
# 2 = pasir lepas
# 3 = material lebih kompak
# 4 = bedrock

litologi = np.zeros_like(X, dtype=int)

litologi[(Z >= 3) & (Z < 10)] = 1
litologi[(Z >= 10) & (Z < bedrock_grid)] = 3
litologi[Z >= bedrock_grid] = 4

# Identifikasi lensa pasir lepas
litologi[lensa1 | lensa2] = 2

nama_litologi = {
    0: "Tanah permukaan/material campuran",
    1: "Sedimen aluvial halus",
    2: "Pasir lepas",
    3: "Material lebih kompak",
    4: "Bedrock"
}


# 5. IDENTIFIKASI KANDIDAT ZONA RENTAN LIKUIFAKSI
# Kriteria penyaringan awal:
# 1. Material berupa pasir lepas
# 2. Berada di bawah muka air tanah
# 3. Vp berada pada rentang indikatif 300-1200 m/s
#
# Kriteria Vp bukan ambang universal untuk likuifaksi.
# Zona dengan Vp lebih tinggi tetap perlu diperiksa apabila
# terdapat pasir jenuh air, karena kejenuhan dapat menaikkan Vp.
#
# Pada simulasi ini, label litologi dan muka air tanah menjadi
# dasar utama, sedangkan Vp digunakan sebagai informasi tambahan.

mask_vp_indikatif = (Vp >= 300) & (Vp <= 1200)
mask_pasir = litologi == 2

kandidat_likuifaksi = (
    mask_pasir
    & mask_di_bawah_air
    & mask_vp_indikatif
)

# Zona pasir jenuh dengan Vp tinggi:
# tetap ditandai untuk pemeriksaan geoteknik, bukan langsung
# diklasifikasikan aman atau berbahaya.
pasir_jenuh_vp_tinggi = (
    mask_pasir
    & mask_di_bawah_air
    & (Vp > 1200)
)

print("\nKRITERIA PENYARINGAN")
print("-" * 65)
print("Material              : pasir lepas (model sintetis)")
print("Kondisi air tanah     : di bawah muka air tanah")
print("Rentang Vp indikatif  : 300-1200 m/s")
print("Vp > 1200 m/s         : tidak otomatis aman; perlu verifikasi")
print("Catatan               : hasil ini bukan klasifikasi risiko final")


# 6. ESTIMASI KETEBALAN SEDIMEN DAN BEDROCK
# Kedalaman bedrock diambil dari model sintetis.
# Pada data nyata, batas ini perlu ditafsirkan dari tomogram,
# data bor, dan informasi geologi.

ketebalan_sedimen = kedalaman_bedrock.copy()

print("\nESTIMASI KETEBALAN SEDIMEN")
print("-" * 65)
print(f"Ketebalan minimum : {ketebalan_sedimen.min():.2f} m")
print(f"Ketebalan maksimum: {ketebalan_sedimen.max():.2f} m")
print(f"Ketebalan rata-rata: {ketebalan_sedimen.mean():.2f} m")


# 7. RINGKASAN KANDIDAT ZONA
luas_satu_sel = jarak_grid * jarak_grid

jumlah_sel_kandidat = int(np.sum(kandidat_likuifaksi))
luas_kandidat = jumlah_sel_kandidat * luas_satu_sel

jumlah_sel_vp_tinggi = int(np.sum(pasir_jenuh_vp_tinggi))
luas_vp_tinggi = jumlah_sel_vp_tinggi * luas_satu_sel

print("\nRINGKASAN ZONA KANDIDAT")
print("-" * 65)
print(f"Jumlah sel kandidat Vp rendah-sedang : {jumlah_sel_kandidat}")
print(f"Luas proyeksi sel kandidat           : {luas_kandidat:.0f} m²")
print(f"Sel pasir jenuh dengan Vp > 1200 m/s : {jumlah_sel_vp_tinggi}")
print(f"Luas proyeksi sel Vp tinggi          : {luas_vp_tinggi:.0f} m²")
print("Luas di atas adalah hasil grid simulasi, bukan luas risiko resmi.")


# 8. VISUALISASI PENAMPANG TOMOGRAFI Vp

fig, ax = plt.subplots(figsize=(14, 6))

gambar = ax.pcolormesh(
    X, Z, Vp,
    shading="auto",
    cmap="turbo",
    vmin=300,
    vmax=2800
)

cbar = plt.colorbar(gambar, ax=ax)
cbar.set_label("Kecepatan gelombang P, Vp (m/s)")

ax.plot(
    x, muka_air,
    color="blue",
    linewidth=2,
    label="Muka air tanah"
)

ax.plot(
    x, kedalaman_bedrock,
    color="black",
    linewidth=2,
    linestyle="--",
    label="Batas bedrock model"
)

ax.set_title("Penampang Tomografi Vp 2D dan Batas Hidrogeologi")
ax.set_xlabel("Jarak lintasan (m)")
ax.set_ylabel("Kedalaman (m)")
ax.invert_yaxis()
ax.legend(loc="lower right")
ax.grid(alpha=0.2)

plt.tight_layout()
plt.savefig("01_tomografi_vp.png", dpi=300, bbox_inches="tight")
plt.show()


# 9. PETA KANDIDAT LIKUIFAKSI

fig, ax = plt.subplots(figsize=(14, 5))

# Latar berupa nilai Vp
gambar = ax.pcolormesh(
    X, Z, Vp,
    shading="auto",
    cmap="gray",
    vmin=300,
    vmax=1800
)

plt.colorbar(
    gambar, ax=ax,
    label="Vp (m/s)"
)

# Overlay zona kandidat
overlay = np.ma.masked_where(
    ~kandidat_likuifaksi,
    kandidat_likuifaksi.astype(float)
)

ax.pcolormesh(
    X, Z, overlay,
    shading="auto",
    cmap=ListedColormap(["red"]),
    alpha=0.85
)

# Overlay pasir jenuh dengan Vp tinggi
overlay_tinggi = np.ma.masked_where(
    ~pasir_jenuh_vp_tinggi,
    pasir_jenuh_vp_tinggi.astype(float)
)

ax.pcolormesh(
    X, Z, overlay_tinggi,
    shading="auto",
    cmap=ListedColormap(["orange"]),
    alpha=0.9
)

ax.plot(x, muka_air, color="blue", linewidth=2)
ax.plot(
    x, kedalaman_bedrock,
    color="black", linestyle="--", linewidth=2
)

legenda = [
    Patch(facecolor="red", label="Kandidat: pasir jenuh, Vp 300-1200 m/s"),
    Patch(facecolor="orange", label="Pasir jenuh, Vp > 1200 m/s: perlu verifikasi")
]

ax.legend(handles=legenda, loc="lower right")
ax.set_title("Overlay Zona Kandidat Likuifaksi")
ax.set_xlabel("Jarak lintasan (m)")
ax.set_ylabel("Kedalaman (m)")
ax.invert_yaxis()
ax.grid(alpha=0.2)

plt.tight_layout()
plt.savefig("02_overlay_kandidat_likuifaksi.png",
            dpi=300, bbox_inches="tight")
plt.show()


# 10. PROFIL KEDALAMAN BEDROCK

plt.figure(figsize=(12, 4))

plt.plot(
    x, kedalaman_bedrock,
    color="black",
    linewidth=2,
    label="Kedalaman bedrock"
)

plt.fill_between(
    x, 0, kedalaman_bedrock,
    alpha=0.25,
    label="Ketebalan sedimen"
)

plt.xlabel("Jarak lintasan (m)")
plt.ylabel("Kedalaman (m)")
plt.title("Profil Kedalaman Bedrock dan Ketebalan Sedimen")
plt.gca().invert_yaxis()
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()

plt.savefig("03_profil_bedrock.png", dpi=300, bbox_inches="tight")
plt.show()


# 11. PROFIL Vp PADA BEBERAPA POSISI

posisi_profil = [50, 150, 250]

plt.figure(figsize=(8, 6))

for posisi in posisi_profil:
    indeks_x = np.argmin(np.abs(x - posisi))

    plt.plot(
        Vp[:, indeks_x],
        z,
        linewidth=2,
        label=f"Posisi x = {x[indeks_x]:.0f} m"
    )

plt.xlabel("Vp (m/s)")
plt.ylabel("Kedalaman (m)")
plt.title("Perbandingan Profil Vp pada Tiga Posisi")
plt.gca().invert_yaxis()
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()

plt.savefig("04_profil_vp.png", dpi=300, bbox_inches="tight")
plt.show()
```

**# 4. STUDI KASUS KETEKNIKAN: DETEKSI RONGGA BAWAH PERMUKAAN (KARST) DENGAN GROUND PENETRATING RADAR (GPR)**

* **Role**
  `Anda berperan sebagai seorang Near-Surface Geophysicist yang menguasai teori propagasi gelombang elektromagnetik (EM), pemrosesan sinyal radar, interpretasi radargram, dan investigasi geofisika untuk kebutuhan infrastruktur transportasi.`

* **Task**
  `Analisis dan interpretasikan data rekaman profil radargram GPR 2D untuk mendeteksi lokasi, kedalaman, serta perkiraan dimensi rongga atau gua karst (void/sinkhole) di bawah permukaan yang berpotensi memicu amblesan atau runtuhan tanah.`

* **Context**
  `Survei GPR ini dilaksanakan di sepanjang koridor perencanaan jalur rel kereta api berkecepatan tinggi yang melintasi kawasan bentang alam karst (batu gamping). Deteksi keberadaan rongga bawah permukaan sangat krusial dilakukan sebelum konstruksi dimulai untuk mencegah risiko kegagalan struktur fondasi rel atau bahaya kecelakaan di masa depan.`

* **Input**
  `Profil radargram GPR 2D (menggunakan frekuensi antena pusat 250 MHz), estimasi kecepatan propagasi gelombang EM pada batu gamping padat (0.1 m/ns), data pola refraksi/refleksi hiperbola, data filter pengolahan (seperti bandpass filter, background removal, dan gain correction), serta peta topografi dan elevasi permukaan.`

* **Constraints**
  `Dilarang menyimpulkan setiap anomali hiperbola pada radargram sebagai rongga karst tanpa memperhatikan fase refleksi (kontras permitivitas relatif). Pertimbangkan efek atenuasi sinyal EM yang tinggi apabila rongga terisi oleh material lempung jenuh atau air asin, yang dapat menyebabkan hilangnya penetrasi sinyal radar (radar signal loss). Bedakan dengan jelas antara respon rongga alami dengan utilitas buatan manusia (seperti pipa atau gorong-gorong).`

* **Output**
  `Rencana output berupa narasi teknis evaluasi geofisika keteknikan yang berisi:`
  `1. Penjelasan mekanisme respon gelombang EM terhadap kontras dielektrik antara batuan gamping dan rongga (berisi udara/air/sebagian lempung).`
  `2. Pemetaan posisi horizontal (jarak lintasan) dan kalkulasi kedalaman anomali hiperbola.`
  `3. Estimasi dimensi rongga berdasarkan geometri kurva hiperbola dan pola refleksi batas bawah rongga.`
  `4. Analisis kedalaman penetrasi efektif sinyal GPR serta faktor pembatas kualitas data.`
  `5. Rekomendasi teknis mitigasi geoteknik (seperti metode injeksi semen/grouting) dan verifikasi drilling atau borehole camera.`

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import butter, sosfiltfilt

np.random.seed(42)

# 1. PARAMETER SURVEI

frekuensi = 250e6       # Hz
v = 0.1                 # m/ns, kecepatan EM pada batu gamping
dx = 0.10               # jarak antartrace (m)
dt = 0.5                # interval sampling waktu (ns)
jarak_maks = 100        # panjang lintasan (m)
waktu_maks = 240        # waktu rekaman (ns)

x = np.arange(0, jarak_maks + dx, dx)
t = np.arange(0, waktu_maks + dt, dt)

X, T = np.meshgrid(x, t)

print("=" * 65)
print("SIMULASI GPR UNTUK DETEKSI RONGGA KARST")
print("=" * 65)
print(f"Frekuensi antena     : {frekuensi/1e6:.0f} MHz")
print(f"Kecepatan EM         : {v:.3f} m/ns")
print(f"Panjang lintasan     : {jarak_maks:.1f} m")
print(f"Interval waktu       : {dt:.2f} ns")
print(f"Jumlah trace         : {len(x)}")
print(f"Jumlah sampel waktu  : {len(t)}")

# 2. FUNGSI WAVELET RICKER

def ricker(tau, f_mhz):
    """
    Wavelet Ricker.
    tau dalam ns; f_mhz dalam MHz.
    """
    a = np.pi * (f_mhz / 1000.0) * tau
    return (1 - 2 * a**2) * np.exp(-a**2)


# 3. MEMBANGUN RADARGRAM SINTETIS
# Target A: kandidat rongga berisi udara
# Posisi horizontal = 42 m
# Kedalaman atap rongga = 3.5 m
# Lebar model = sekitar 4 m
#
# Target B: objek menyerupai pipa/gorong-gorong
# Posisi horizontal = 72 m
# Kedalaman pusat = 1.5 m
#
# Semua posisi ini adalah asumsi simulasi.

radar_raw = np.zeros((len(t), len(x)))

# Reflektor lapisan dangkal
for kedalaman, amplitudo in [
    (0.5, 0.45),
    (1.4, 0.28),
    (5.5, 0.18),
    (8.0, 0.12)
]:
    t0 = 2 * kedalaman / v
    wave = ricker(t - t0, 250)
    radar_raw += amplitudo * wave[:, None]

# 4. RESPON HIPERBOLA KANDIDAT RONGGA

x_void = 42.0
z_void = 3.5
lebar_void_model = 4.0

t0_void = 2 * z_void / v

t_hyperbola_void = np.sqrt(
    t0_void**2
    + (2 * (X - x_void) / v)**2
)

# Difraksi dari tepi rongga
radar_raw += 0.85 * ricker(T - t_hyperbola_void, 250)

# Refleksi batas atas rongga dengan kontras dielektrik kuat
# Refleksi bawah dimodelkan secara lebih lemah.
z_void_bottom = 5.0
t_void_bottom = 2 * z_void_bottom / v

mask_void = np.abs(X - x_void) <= lebar_void_model / 2

radar_raw += (
    mask_void
    * 0.55
    * ricker(T - t0_void, 250)
)

radar_raw += (
    mask_void
    * -0.22
    * ricker(T - t_void_bottom, 250)
)

# 5. RESPON OBJEK BUATAN: PIPA/GORONG-GORONG

x_pipe = 72.0
z_pipe = 1.5
t0_pipe = 2 * z_pipe / v

t_hyperbola_pipe = np.sqrt(
    t0_pipe**2
    + (2 * (X - x_pipe) / v)**2
)

radar_raw += 0.65 * ricker(T - t_hyperbola_pipe, 250)

# 6. ZONA ATENUASI TINGGI
# Simulasi material lempung jenuh yang meredam sinyal.
# Ini bukan model elektromagnetik penuh.

zona_atenuasi = (
    (X >= 83) & (X <= 94)
    & (T >= 40)
)

faktor_atenuasi = np.ones_like(radar_raw)
faktor_atenuasi[zona_atenuasi] = 0.12

radar_raw *= faktor_atenuasi

# Tambahkan tren latar dan noise
tren_latar = (
    0.06 * np.sin(2 * np.pi * T / 80)
    + 0.04 * np.cos(2 * np.pi * X / 35)
)

noise = np.random.normal(0, 0.10, size=radar_raw.shape)

radar_raw += tren_latar + noise


# 7. PEMROSESAN DATA GPR

# 7a. Background removal
# Mengurangi rata-rata tiap sampel waktu di seluruh lintasan.
background = np.mean(radar_raw, axis=1, keepdims=True)
radar_bg = radar_raw - background

# 7b. Bandpass filter
# Rentang ilustratif 80-500 MHz.
# Frekuensi sampling = 1/dt dalam GHz.
fs_ghz = 1.0 / dt
nyquist = fs_ghz / 2.0

low_cut = 0.08   # GHz = 80 MHz
high_cut = 0.50  # GHz = 500 MHz

if high_cut >= nyquist:
    high_cut = 0.95 * nyquist

sos = butter(
    4,
    [low_cut / nyquist, high_cut / nyquist],
    btype="bandpass",
    output="sos"
)

radar_filtered = sosfiltfilt(sos, radar_bg, axis=0)

# 7c. Gain correction
# Gain eksponensial untuk menonjolkan refleksi yang lebih dalam.
gain = np.exp(T / 105.0)
radar_processed = radar_filtered * gain

# Normalisasi robust untuk tampilan
skala = np.percentile(np.abs(radar_processed), 99.5)
radar_display = np.clip(radar_processed / skala, -1, 1)

print("\nPEMROSESAN")
print("-" * 65)
print("1. Background removal")
print("2. Bandpass filter 80-500 MHz (rentang ilustratif)")
print("3. Gain correction eksponensial")
print("Catatan: parameter filter perlu disesuaikan dengan data lapangan.")


# 8. VISUALISASI RADARGRAM SEBELUM DAN SESUDAH PEMROSESAN

fig, axes = plt.subplots(1, 2, figsize=(16, 6), sharey=True)

axes[0].imshow(
    radar_raw,
    aspect="auto",
    extent=[x.min(), x.max(), t.max(), t.min()],
    cmap="gray",
    vmin=-0.8,
    vmax=0.8
)
axes[0].set_title("Radargram Sebelum Pemrosesan")
axes[0].set_xlabel("Jarak lintasan (m)")
axes[0].set_ylabel("Waktu tempuh dua arah (ns)")

axes[1].imshow(
    radar_display,
    aspect="auto",
    extent=[x.min(), x.max(), t.max(), t.min()],
    cmap="gray",
    vmin=-0.8,
    vmax=0.8
)
axes[1].set_title("Radargram Setelah Pemrosesan")
axes[1].set_xlabel("Jarak lintasan (m)")

for ax in axes:
    ax.axvline(x_void, color="red", linestyle="--", alpha=0.8)
    ax.axvline(x_pipe, color="blue", linestyle="--", alpha=0.8)
    ax.grid(alpha=0.15)

plt.tight_layout()
plt.savefig("01_radargram_raw_processed.png", dpi=300,
            bbox_inches="tight")
plt.show()


# 9. ESTIMASI KEDALAMAN DARI PUNCAK HIPERBOLA
# Pada data nyata, waktu puncak harus dipetik dari radargram
# setelah koreksi time-zero dan pemilihan kecepatan yang tepat.

def hitung_kedalaman(twtt_ns, velocity_m_per_ns):
    return velocity_m_per_ns * twtt_ns / 2.0

kedalaman_void = hitung_kedalaman(t0_void, v)
kedalaman_pipe = hitung_kedalaman(t0_pipe, v)

print("\nHASIL ESTIMASI KEDALAMAN")
print("-" * 65)
print(f"Kandidat rongga: x = {x_void:.1f} m")
print(f"Waktu puncak   : {t0_void:.1f} ns")
print(f"Kedalaman atap : {kedalaman_void:.2f} m")
print()
print(f"Objek pipa     : x = {x_pipe:.1f} m")
print(f"Waktu puncak   : {t0_pipe:.1f} ns")
print(f"Kedalaman pusat: {kedalaman_pipe:.2f} m")


# 10. ESTIMASI KECEPATAN DARI FITTING HIPERBOLA
# Bentuk hiperbola: t(x)^2 = t0^2 + 4(x-x0)^2/v^2
# Jika x0 diketahui atau diperkirakan, slope regresi t^2
# terhadap (x-x0)^2 dapat digunakan untuk estimasi v.

def estimasi_kecepatan_hiperbola(
    xdata, tdata, x0, t0, rentang_x=8
):
    mask = (
        (np.abs(xdata - x0) <= rentang_x)
        & (tdata >= t0)
    )

    xx = (xdata[mask] - x0) ** 2
    tt = tdata[mask] ** 2

    if len(xx) < 5:
        return np.nan

    # t^2 - t0^2 = (4/v^2) * (x-x0)^2
    yy = tt - t0**2

    slope = np.sum(xx * yy) / np.sum(xx**2)

    if slope <= 0:
        return np.nan

    return 2 / np.sqrt(slope)

# Data teoritis untuk menunjukkan cara fitting.
# Pada survei nyata, gunakan waktu puncak hasil picking.
xfit = np.linspace(x_void - 8, x_void + 8, 100)
tfit = np.sqrt(
    t0_void**2 + (2 * (xfit - x_void) / v)**2
)

v_estimasi = estimasi_kecepatan_hiperbola(
    xfit, tfit, x_void, t0_void, rentang_x=8
)

print("\nESTIMASI KECEPATAN HIPERBOLA")
print("-" * 65)
print(f"Kecepatan model     : {v:.3f} m/ns")
print(f"Kecepatan hasil fit : {v_estimasi:.3f} m/ns")
print("Nilai ini menggunakan kurva teoritis, bukan picking data nyata.")


# 11. VISUALISASI HIPERBOLA DAN KEDALAMAN

fig, ax = plt.subplots(figsize=(13, 6))

ax.imshow(
    radar_display,
    aspect="auto",
    extent=[x.min(), x.max(), t.max(), t.min()],
    cmap="gray",
    vmin=-0.8,
    vmax=0.8
)

# Overlay kurva teoritis kandidat rongga
ax.plot(
    x,
    np.sqrt(t0_void**2 + (2 * (x - x_void) / v)**2),
    color="red",
    linewidth=2,
    label="Kurva hiperbola kandidat rongga"
)

# Overlay kurva teoritis objek pipa
ax.plot(
    x,
    np.sqrt(t0_pipe**2 + (2 * (x - x_pipe) / v)**2),
    color="cyan",
    linewidth=2,
    label="Kurva hiperbola objek menyerupai pipa"
)

ax.scatter(
    [x_void, x_pipe],
    [t0_void, t0_pipe],
    color=["red", "cyan"],
    edgecolor="black",
    s=70,
    zorder=5
)

ax.set_title("Pemetaan Hiperbola dan Posisi Anomali")
ax.set_xlabel("Jarak lintasan (m)")
ax.set_ylabel("Waktu tempuh dua arah (ns)")
ax.legend(loc="lower right")
ax.grid(alpha=0.15)

plt.tight_layout()
plt.savefig("02_picking_hiperbola.png", dpi=300,
            bbox_inches="tight")
plt.show()


# 12. ESTIMASI DIMENSI RONGGA
# Lebar lateral tidak dapat dihitung secara unik hanya dari
# lebar hiperbola. Di sini, lebar target diketahui dari model.
# Pada data nyata, batas atap/bawah perlu ditelusuri dan
# dikonfirmasi menggunakan lintasan silang serta pemboran.

lebar_void_estimasi = lebar_void_model
kedalaman_atap = kedalaman_void
kedalaman_dasar = z_void_bottom
tinggi_void_estimasi = kedalaman_dasar - kedalaman_atap

print("\nESTIMASI GEOMETRI KANDIDAT RONGGA")
print("-" * 65)
print(f"Posisi horizontal   : {x_void:.1f} m")
print(f"Kedalaman atap      : {kedalaman_atap:.2f} m")
print(f"Kedalaman dasar     : {kedalaman_dasar:.2f} m")
print(f"Tinggi model        : {tinggi_void_estimasi:.2f} m")
print(f"Lebar model         : {lebar_void_estimasi:.2f} m")
print("PENTING: lebar dan tinggi ini diketahui dari model sintetis.")
print("Pada data nyata, dimensi rongga perlu diperkirakan dengan")
print("refleksi batas, pemodelan, lintasan silang, dan verifikasi bor.")

# 13. ANALISIS PENETRASI DAN PANJANG GELOMBANG

panjang_gelombang = v / (frekuensi / 1e9)

print("\nPARAMETER RESOLUSI DAN PENETRASI")
print("-" * 65)
print(f"Panjang gelombang       : {panjang_gelombang:.3f} m")
print(f"Seperempat panjang gelombang: {panjang_gelombang/4:.3f} m")
print("Angka tersebut adalah pendekatan resolusi vertikal ideal.")
print("Resolusi aktual dipengaruhi bandwidth, noise, atenuasi,")
print("kontras dielektrik, dan geometri target.")

# 14. SKEMA KONSEPTUAL POSISI TARGET

fig, ax = plt.subplots(figsize=(14, 4))

ax.fill_between(
    [0, 100],
    [0, 0],
    [10, 10],
    color="tan",
    alpha=0.35,
    label="Zona batu gamping (skema)"
)

# Kandidat rongga
rect_void = plt.Rectangle(
    (x_void - lebar_void_model / 2, z_void),
    lebar_void_model,
    tinggi_void_estimasi,
    facecolor="white",
    edgecolor="red",
    linewidth=2,
    label="Kandidat rongga karst"
)
ax.add_patch(rect_void)

# Objek pipa sebagai lingkaran
pipe = plt.Circle(
    (x_pipe, z_pipe),
    0.45,
    facecolor="lightblue",
    edgecolor="blue",
    linewidth=2,
    label="Objek menyerupai pipa"
)
ax.add_patch(pipe)

ax.set_xlim(0, 100)
ax.set_ylim(10, 0)
ax.set_xlabel("Jarak lintasan (m)")
ax.set_ylabel("Kedalaman (m)")
ax.set_title("Skema Konseptual Kandidat Rongga dan Objek Buatan")
ax.legend(loc="lower right")
ax.grid(alpha=0.2)

plt.tight_layout()
plt.savefig("03_skema_kandidat_rongga.png", dpi=300,
            bbox_inches="tight")
plt.show()
```

**# 5. STUDI KASUS HIDRO-GEOLOGI LEPAS PANTAI: PEMETAAN AIR TANAH LEPAS PANTAI (OFFSHORE AQUIFER) DENGAN METODE ELECTROMAGNETIC (CSEM)**

* **Role**
  `Anda berperan sebagai seorang Marine Geophysicist Senior yang ahli dalam metode Controlled-Source Electromagnetic (CSEM) bawah laut, pemodelan inversi konduktivitas/resistivitas media fluida laut, serta hidrogeologi kawasan pesisir dan lepas pantai.`

* **Task**
  `Lakukan analisis dan evaluasi terhadap data resistivitas 2D/3D hasil pemodelan CSEM dasar laut untuk mengidentifikasi keberadaan, geometri, dan perkiraan batas sebaran lateral serta vertikal dari akuifer air tawar bawah laut (offshore freshened groundwater).`

* **Context**
  `Penelitian dilakukan di kawasan pesisir pulau kecil yang mengalami krisis air bersih dan penurunan muka air tanah yang parah akibat eksploitasi berlebih. Pengembang dan pemerintah daerah berupaya mencari sumber daya air tawar non-konvensional yang tersimpan di dalam lapisan sedimen di bawah dasar laut (lepas pantai) sebagai cadangan strategis masa depan.`

* **Input**
  `Data komponen medan listrik dan magnetik hasil perekaman transmitter-receiver CSEM di dasar laut, profil penampang inversi resistivitas bawah dasar laut (kedalaman 0 - 500 meter di bawah sediment-water interface), data kedalaman air laut (batimetri), data nilai resistivitas air laut lokal (0.3 ohm.m), serta data seismik dangkal pencitraan struktur sedimen jika tersedia.`

* **Constraints**
  `Dalam analisis, perhatikan dengan teliti bahwa air laut dan sedimen jenuh air laut memiliki nilai resistivitas yang sangat rendah (< 1 ohm.m). Oleh karena itu, keberadaan akuifer air tawar akan ditandai oleh kontras resistivitas yang lebih tinggi (biasanya > 5 - 10 ohm.m). Namun, jangan mengabaikan kemungkinan bahwa respon resistivitas tinggi tersebut berasal dari batuan karbonat, evaporit, atau lapisan gas dangkal (shallow gas). Analisis wajib memisahkan respon akibat jenis fluida dengan jenis batuan matrik.`

* **Output**
  `Rencana output berupa narasi analisis geofisika kelautan yang detail dan ilmiah, memuat:`
  `1. Prinsip kerja difusi gelombang EM di media air laut dan penetrasinya ke dalam sedimen dasar laut.`
  `2. Identifikasi dan pemetaan batas zona resistivitas tinggi yang diindikasikan sebagai akuifer air tawar lepas pantai.`
  `3. Estimasi kedalaman perlapisan akuifer di bawah dasar laut beserta perkiraan ketebalannya.`
  `4. Diskusi mendalam mengenai ambiguitas inversi CSEM, efek air-wave, dan perbedaan respon fluida vs litologi.`
  `5. Kesimpulan mengenai potensi cadangan air serta saran investigasi lanjutan melalui pemboran geotechnical offshore dan pengujian sampel geokimia fluida.`

```python

import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import gaussian_filter

np.random.seed(42)

# 1. PARAMETER SURVEI

rho_air_laut = 0.3       # Resistivitas air laut (Ohm.m)
frekuensi = 0.25         # Frekuensi CSEM (Hz)
dx = 50.0                # Interval horizontal (m)
dz = 5.0                 # Interval vertikal (m)
panjang_lintasan = 6000  # Panjang lintasan (m)
kedalaman_maks = 500     # Kedalaman di bawah dasar laut (m)

x = np.arange(0, panjang_lintasan + dx, dx)
z = np.arange(0, kedalaman_maks + dz, dz)

X, Z = np.meshgrid(x, z)

print("=" * 65)
print("SIMULASI MARINE CSEM UNTUK PEMETAAN AKUIFER LEPAS PANTAI")
print("=" * 65)
print(f"Resistivitas air laut : {rho_air_laut:.2f} Ohm.m")
print(f"Frekuensi CSEM        : {frekuensi:.2f} Hz")
print(f"Panjang lintasan      : {panjang_lintasan:.0f} m")
print(f"Kedalaman maksimum    : {kedalaman_maks:.0f} m")
print(f"Interval horizontal   : {dx:.0f} m")
print(f"Interval vertikal     : {dz:.0f} m")
print(f"Jumlah titik horizontal: {len(x)}")
print(f"Jumlah titik vertikal  : {len(z)}")

# 2. MEMBANGUN MODEL RESISTIVITAS BAWAH DASAR LAUT
# Seluruh kedalaman z dihitung dari dasar laut.
#
# Target A: kandidat akuifer air tawar
# Posisi horizontal : 2200-4000 m
# Kedalaman         : 120-220 m
# Resistivitas      : 8 Ohm.m
#
# Target B: batuan resistif, bukan akuifer
# Posisi horizontal : 4400-5100 m
# Kedalaman         : 240-330 m
# Resistivitas      : 15 Ohm.m
#
# Target C: zona dangkal dengan resistivitas menengah
# Posisi horizontal : 700-1300 m
# Kedalaman         : 45-90 m
# Resistivitas      : 5.5 Ohm.m
#
# Semua nilai adalah asumsi simulasi.

rho_model = np.full(
    (len(z), len(x)),
    0.8,
    dtype=float
)

# Target A: kandidat akuifer
x_aquifer_min = 2200
x_aquifer_max = 4000
z_aquifer_top = 120
z_aquifer_bottom = 220
rho_aquifer = 8.0

mask_aquifer = (
    (X >= x_aquifer_min)
    & (X <= x_aquifer_max)
    & (Z >= z_aquifer_top)
    & (Z <= z_aquifer_bottom)
)

rho_model[mask_aquifer] = rho_aquifer

# Target B: batuan resistif sebagai sumber ambiguitas
mask_batuan = (
    (X >= 4400)
    & (X <= 5100)
    & (Z >= 240)
    & (Z <= 330)
)

rho_model[mask_batuan] = 15.0

# Target C: anomali resistivitas dangkal
mask_dangkal = (
    (X >= 700)
    & (X <= 1300)
    & (Z >= 45)
    & (Z <= 90)
)

rho_model[mask_dangkal] = 5.5

# Penghalusan model untuk visualisasi
rho_model_halus = gaussian_filter(
    rho_model,
    sigma=(1.2, 1.2)
)

# 3. FUNGSI KEDALAMAN SKIN DEPTH

def hitung_skin_depth(rho, f):
    return 503.0 * np.sqrt(rho / f)


delta_air_laut = hitung_skin_depth(
    rho_air_laut,
    frekuensi
)

delta_sedimen = hitung_skin_depth(
    0.8,
    frekuensi
)

delta_akuifer = hitung_skin_depth(
    rho_aquifer,
    frekuensi
)

print("\nESTIMASI SKIN DEPTH")
print("-" * 65)
print(f"Air laut              : {delta_air_laut:.1f} m")
print(f"Sedimen latar         : {delta_sedimen:.1f} m")
print(f"Kandidat akuifer      : {delta_akuifer:.1f} m")
print("Catatan: skin depth bukan kedalaman investigasi pasti.")

# 3. ILUSTRASI RESPONS MARINE CSEM

offset = np.arange(500, 6001, 250)

rho_latar = 0.8
kontras_rho = (
    rho_model_halus - rho_latar
) / rho_latar

# Indeks posisi transmitter ilustratif
x_tx = 3000.0
indeks_tx = np.argmin(np.abs(x - x_tx))

# Profil resistivitas vertikal di sekitar transmitter
profil_tx = rho_model_halus[:, indeks_tx]

respons_offset = np.zeros(len(offset))

for i, r in enumerate(offset):
    # Faktor atenuasi offset yang disederhanakan
    atenuasi_offset = np.exp(
        -r / (2.0 * delta_sedimen)
    )

    # Kontras resistivitas pada model
    kontras_rata = np.mean(
        kontras_rho[
            (Z[:, indeks_tx] >= 50)
            & (Z[:, indeks_tx] <= 300),
            indeks_tx
        ]
    )

    respons_offset[i] = (
        atenuasi_offset * kontras_rata
    )

# Respons relatif untuk visualisasi
respons_maks = np.max(np.abs(respons_offset))

if respons_maks > 0:
    respons_normal = respons_offset / respons_maks
else:
    respons_normal = respons_offset.copy()

# 5. MEMBUAT DATA SINTETIS DAN MODEL HASIL INVERSI SEMU

noise_model = np.random.normal(
    loc=0.0,
    scale=0.20,
    size=rho_model_halus.shape
)

rho_observasi_semu = (
    rho_model_halus
    * np.exp(noise_model)
)

rho_inversi_semu = gaussian_filter(
    rho_observasi_semu,
    sigma=(3.0, 3.0)
)

rho_inversi_semu = np.clip(
    rho_inversi_semu,
    0.2,
    30.0
)

print("\nPEMBENTUKAN MODEL SINTETIS")
print("-" * 65)
print("1. Model resistivitas awal dibuat.")
print("2. Noise sintetis ditambahkan.")
print("3. Penghalusan digunakan untuk meniru efek resolusi.")
print("4. Model halus bukan hasil inversi elektromagnetik penuh.")

# 4. VISUALISASI MODEL RESISTIVITAS

fig, ax = plt.subplots(figsize=(14, 6))

im = ax.pcolormesh(
    X,
    Z,
    rho_model_halus,
    shading="auto",
    cmap="turbo",
    vmin=0.5,
    vmax=15
)

cbar = plt.colorbar(im, ax=ax)
cbar.set_label("Resistivitas (Ohm.m)")

ax.axvline(
    x_tx,
    color="white",
    linestyle="--",
    linewidth=1.5,
    label="Posisi transmitter ilustratif"
)

ax.set_title(
    "Model Resistivitas Bawah Dasar Laut"
)
ax.set_xlabel("Jarak lintasan (m)")
ax.set_ylabel("Kedalaman di bawah dasar laut (m)")
ax.set_ylim(kedalaman_maks, 0)
ax.legend(loc="lower right")
ax.grid(alpha=0.15)

plt.tight_layout()
plt.savefig(
    "01_model_resistivitas_CSEM.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()

# 5. PROFIL RESISTIVITAS VERTIKAL

posisi_profil = [1000, 3000, 4700]

fig, ax = plt.subplots(figsize=(8, 7))

for posisi in posisi_profil:
    indeks = np.argmin(np.abs(x - posisi))

    ax.plot(
        rho_model_halus[:, indeks],
        z,
        linewidth=2,
        label=f"x = {x[indeks]:.0f} m"
    )

ax.set_title("Profil Resistivitas Vertikal")
ax.set_xlabel("Resistivitas (Ohm.m)")
ax.set_ylabel("Kedalaman di bawah dasar laut (m)")
ax.set_ylim(kedalaman_maks, 0)
ax.grid(alpha=0.3)
ax.legend()

plt.tight_layout()
plt.savefig(
    "02_profil_resistivitas_CSEM.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()

# 8. ILUSTRASI RESPONS CSEM TERHADAP OFFSET

fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(
    offset,
    respons_normal,
    marker="o",
    linewidth=2,
    label="Respons relatif sintetis"
)

ax.set_title(
    "Ilustrasi Respons Relatif Marine CSEM"
)
ax.set_xlabel("Offset transmitter-receiver (m)")
ax.set_ylabel("Respons relatif (tanpa satuan)")
ax.grid(alpha=0.3)
ax.legend()

plt.tight_layout()
plt.savefig(
    "03_respons_offset_CSEM.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()

# 6. ESTIMASI KEDALAMAN DAN KETEBALAN KANDIDAT AKUIFER

ambang_resistivitas = 5.0

mask_kandidat = (
    (rho_model_halus >= ambang_resistivitas)
    & (X >= 1800)
    & (X <= 4200)
    & (Z >= 50)
    & (Z <= 300)
)

baris, kolom = np.where(mask_kandidat)

print("\nESTIMASI ZONA KANDIDAT AKUIFER")
print("-" * 65)

if len(baris) > 0:
    x_min_estimasi = x[kolom].min()
    x_max_estimasi = x[kolom].max()
    z_min_estimasi = z[baris].min()
    z_max_estimasi = z[baris].max()

    ketebalan_estimasi = (
        z_max_estimasi - z_min_estimasi + dz
    )

    print(
        f"Rentang horizontal : "
        f"{x_min_estimasi:.0f}-{x_max_estimasi:.0f} m"
    )
    print(
        f"Kedalaman atas     : "
        f"{z_min_estimasi:.0f} m"
    )
    print(
        f"Kedalaman bawah    : "
        f"{z_max_estimasi:.0f} m"
    )
    print(
        f"Ketebalan indikatif: "
        f"{ketebalan_estimasi:.0f} m"
    )
else:
    print("Tidak ditemukan sel yang memenuhi ambang.")

print(
    "Catatan: hasil bergantung pada ambang resistivitas "
    "dan model sintetis."
)

# 10. PEMETAAN ZONA KANDIDAT AKUIFER

fig, ax = plt.subplots(figsize=(14, 6))

im = ax.pcolormesh(
    X,
    Z,
    rho_model_halus,
    shading="auto",
    cmap="turbo",
    vmin=0.5,
    vmax=15
)

plt.colorbar(
    im,
    ax=ax,
    label="Resistivitas (Ohm.m)"
)

ax.contour(
    X,
    Z,
    rho_model_halus,
    levels=[ambang_resistivitas],
    colors="white",
    linewidths=2
)

ax.contourf(
    X,
    Z,
    mask_kandidat.astype(float),
    levels=[0.5, 1.5],
    colors=["none", "lime"],
    alpha=0.30
)

ax.set_title(
    "Peta Zona Resistif Kandidat Akuifer"
)
ax.set_xlabel("Jarak lintasan (m)")
ax.set_ylabel("Kedalaman di bawah dasar laut (m)")
ax.set_ylim(kedalaman_maks, 0)
ax.grid(alpha=0.15)

plt.tight_layout()
plt.savefig(
    "04_peta_kandidat_akuifer_CSEM.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()

# 7. PERBANDINGAN MODEL AWAL DAN MODEL HALUS

fig, axes = plt.subplots(
    1, 2,
    figsize=(15, 6),
    sharey=True
)

im1 = axes[0].pcolormesh(
    X,
    Z,
    rho_model,
    shading="auto",
    cmap="turbo",
    vmin=0.5,
    vmax=15
)

axes[0].set_title("Model Resistivitas Awal")
axes[0].set_xlabel("Jarak lintasan (m)")
axes[0].set_ylabel("Kedalaman di bawah dasar laut (m)")

im2 = axes[1].pcolormesh(
    X,
    Z,
    rho_inversi_semu,
    shading="auto",
    cmap="turbo",
    vmin=0.5,
    vmax=15
)

axes[1].set_title("Model Halus dengan Noise")
axes[1].set_xlabel("Jarak lintasan (m)")

fig.colorbar(
    im2,
    ax=axes,
    label="Resistivitas (Ohm.m)",
    shrink=0.85
)

for ax in axes:
    ax.set_ylim(kedalaman_maks, 0)
    ax.grid(alpha=0.15)

plt.suptitle(
    "Perbandingan Model Awal dan Model Halus Sintetis"
)
plt.tight_layout()
plt.savefig(
    "05_perbandingan_model_CSEM.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()

# 8. ILUSTRASI PENAMPANG 3D SEMU

from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure(figsize=(13, 7))
ax = fig.add_subplot(111, projection="3d")

permukaan = ax.plot_surface(
    X,
    Z,
    rho_model_halus,
    cmap="turbo",
    linewidth=0,
    antialiased=True
)

fig.colorbar(
    permukaan,
    ax=ax,
    shrink=0.65,
    pad=0.10,
    label="Resistivitas (Ohm.m)"
)

ax.set_title(
    "Penampang 3D Semu Resistivitas Bawah Dasar Laut"
)
ax.set_xlabel("Jarak lintasan (m)")
ax.set_ylabel("Kedalaman (m)")
ax.set_zlabel("Resistivitas (Ohm.m)")
ax.invert_yaxis()

plt.tight_layout()
plt.savefig(
    "06_penampang_3D_semu_CSEM.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()

# 9. ANALISIS AMBIGUITAS RESISTIVITAS

fig, axes = plt.subplots(
    1, 2,
    figsize=(14, 5),
    sharey=True
)

im_a = axes[0].pcolormesh(
    X,
    Z,
    rho_model_halus,
    shading="auto",
    cmap="turbo",
    vmin=0.5,
    vmax=15
)

axes[0].set_title(
    "Zona Resistif: Kandidat Akuifer"
)
axes[0].set_xlabel("Jarak lintasan (m)")
axes[0].set_ylabel("Kedalaman (m)")
axes[0].set_xlim(1800, 4200)
axes[0].set_ylim(300, 0)

im_b = axes[1].pcolormesh(
    X,
    Z,
    rho_model_halus,
    shading="auto",
    cmap="turbo",
    vmin=0.5,
    vmax=15
)

axes[1].set_title(
    "Zona Resistif: Kemungkinan Batuan"
)
axes[1].set_xlabel("Jarak lintasan (m)")
axes[1].set_xlim(4200, 5400)
axes[1].set_ylim(350, 0)

fig.colorbar(
    im_b,
    ax=axes,
    label="Resistivitas (Ohm.m)",
    shrink=0.85
)

plt.suptitle(
    "Ambiguitas Interpretasi Resistivitas"
)
plt.tight_layout()
plt.savefig(
    "07_ambiguitas_resistivitas_CSEM.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```
